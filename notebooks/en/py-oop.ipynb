A class is a blueprint for objects and keeps data (attributes) and behavior (methods) together. `__init__` runs when an object is created, and `self` refers to that object.

scikit-learn models are classes too, which is why we write `model.fit(X, y)`.

## Four main ideas
| Idea | Meaning |
|---|---|
| Encapsulation | Data and behavior together |
| Inheritance | A child class reuses a parent class |
| Polymorphism | Same method name, different behavior |
| Abstraction | Users see only the interface |

The `__str__` method controls how an object prints.



```python
class Shape:
    def area(self):
        raise NotImplementedError
    def __str__(self):
        return f"{type(self).__name__} area = {self.area():.2f}"

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14159 * self.r ** 2

class Square(Shape):
    def __init__(self, a):
        self.a = a
    def area(self):
        return self.a ** 2

for s in [Circle(1), Square(2)]:
    print(s)
```

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        self.balance += amount
        return self.balance

a = Account("Mina")
a.deposit(50)
print(a.owner, a.balance)

class Savings(Account):
    def add_interest(self, rate):
        return self.deposit(self.balance * rate)

s = Savings("Omid", 100)
print(s.add_interest(0.1))
```
