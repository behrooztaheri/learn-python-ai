# Regression metrics

A lesson from the Python and AI School, by Dr. Behrooz Taheri

$$MAE=\frac{1}{n}\sum|y_i-\hat y_i|\qquad RMSE=\sqrt{\frac{1}{n}\sum(y_i-\hat y_i)^2}$$

$$R^2=1-\frac{\sum(y_i-\hat y_i)^2}{\sum(y_i-\bar y)^2}$$

- **MAE** is the average error size, in output units.
- **RMSE** is more sensitive to large errors.
- **R²** is the share of variation explained, and it can be negative.

> R² of 0.9 means 90% of the variation is explained, not that 90% of predictions are right.

## Worked example with four houses
| House | Actual y | Predicted ŷ | Error | Absolute | Square |
|---|---|---|---|---|---|
| 1 | 100 | 110 | −10 | 10 | 100 |
| 2 | 150 | 140 | 10 | 10 | 100 |
| 3 | 200 | 210 | −10 | 10 | 100 |
| 4 | 250 | 240 | 10 | 10 | 100 |

- Mean absolute error: $MAE=40/4=10$
- Mean squared error: $MSE=400/4=100$
- Its root: $RMSE=\sqrt{100}=10$
- The actual mean is 175, so $\sum(y-\bar y)^2=12500$ and $R^2=1-400/12500=0.968$

## Which metric?
| Metric | Better | Unit | When |
|---|---|---|---|
| MAE | Near 0 | Output unit | Simple error statement |
| MSE | Near 0 | Squared unit | Training loss |
| RMSE | Near 0 | Output unit | Large errors are costly |
| R² | Near 1 | None | Comparing with predicting the mean |
| MAPE | Near 0 | Percent | Outputs far from zero |

> A negative R² means the model is worse than simply predicting the mean.

```python
import numpy as np

y = np.array([100, 150, 200, 250])
p = np.array([110, 140, 210, 240])
e = y - p

print("MAE :", np.mean(np.abs(e)))
print("RMSE:", np.sqrt(np.mean(e ** 2)))
print("R2  :", 1 - np.sum(e ** 2) / np.sum((y - y.mean()) ** 2))
```
