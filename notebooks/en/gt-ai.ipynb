## Shapley value
A player's share is their average added value over all orders of arrival.



```python
from itertools import permutations

v = {(): 0, ("A",): 10, ("B",): 20, ("C",): 30, ("A", "B"): 40, ("A", "C"): 50, ("B", "C"): 60, ("A", "B", "C"): 90}
val = lambda s: v[tuple(sorted(s))]

shap = dict.fromkeys("ABC", 0.0)
orders = list(permutations("ABC"))
for order in orders:
    seen = []
    for p in order:
        shap[p] += val(seen + [p]) - val(seen)
        seen.append(p)
print({k: round(s / len(orders), 2) for k, s in shap.items()})
```
