Every Python file is a module. `import` gives access to existing code. The standard library comes with Python, for example `math`, `random`, `datetime` and `collections`.

The block `if __name__ == "__main__":` runs only when the file is executed directly, not when it is imported.

```python
import math
import random
from collections import Counter
from datetime import date

print(math.sqrt(144), round(math.pi, 4))
random.seed(1)
print(random.randint(1, 6))
print(Counter("mississippi").most_common(2))
print(date(2026, 1, 1).strftime("%A"))
```
