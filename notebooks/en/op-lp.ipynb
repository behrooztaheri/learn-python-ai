In linear programming, the objective and constraints are linear. The best solution lies at a corner of the feasible region.

## Example
A factory makes two products and wants maximum profit:

$$\max\ 3x+5y\quad s.t.\ x\le4,\ 2y\le12,\ 3x+2y\le18,\ x,y\ge0$$

| Corner | Profit $3x+5y$ |
|---|---|
| (0,0) | 0 |
| (4,0) | 12 |
| (4,3) | 27 |
| (2,6) | **36** |
| (0,6) | 30 |



```python
from scipy.optimize import linprog

result = linprog(c=[-3, -5], A_ub=[[1, 0], [0, 2], [3, 2]], b_ub=[4, 12, 18], bounds=[(0, None), (0, None)])
print("x, y =", result.x.round(3), "profit =", round(-result.fun, 3))
```



> `linprog` minimizes, so negate the objective to maximize.
