## macOS and Linux
> On macOS and Linux, use `python3` instead of `python` and `pip3` instead of `pip`.

```bash
# Windows
py install
python --version
py list

# macOS and Linux
python3 --version
```

```bash
# Ubuntu / Debian
sudo apt update
sudo apt install python3 python3-pip python3-venv
```
