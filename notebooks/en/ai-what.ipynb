# What is AI?

A lesson from the Python and AI School, by Dr. Behrooz Taheri

**Artificial intelligence** is the field of building systems that perform tasks requiring intelligence. **Machine learning** is a part of it that learns patterns from data instead of hand-written rules. **Deep learning** is a part of machine learning that uses multi-layer neural networks.

## Uses across fields
- Medicine: supporting diagnosis from images
- Agriculture: detecting plant disease
- Finance: spotting fraud
- Language: translation and chat assistants

## When is it a poor fit?
If a simple rule is enough, data is scarce, or errors are costly and explanations are required, decide with care.
