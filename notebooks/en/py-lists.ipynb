A list is an ordered, changeable collection. Positions start at zero and negative positions count from the end.

A **list comprehension** is a short way to build a new list from another one.

## List methods
| Method | Purpose | Changes the list? |
|---|---|---|
| `append(x)` | Add at the end | Yes |
| `insert(i, x)` | Add at position i | Yes |
| `pop(i)` | Remove and return | Yes |
| `sort()` | Sort in place | Yes |
| `sorted(lst)` | New sorted list | No |
| `count(x)` | Count | No |

> A common mistake is changing a list while looping over it. Loop over a copy or build a new list.



```python
nums = [1, 2, 2, 3, 2]

result = [n for n in nums if n != 2]
print(result)

matrix = [[1, 2, 3], [4, 5, 6]]
print([row[1] for row in matrix])
print([x for row in matrix for x in row])
```

```python
nums = [5, 2, 9, 1]
nums.append(7)
nums.remove(2)
nums.sort()
print(nums, nums[-1], nums[1:3])
print(len(nums), sum(nums), max(nums))

squares = [n ** 2 for n in nums]
evens = [n for n in range(10) if n % 2 == 0]
print(squares, evens)
```
