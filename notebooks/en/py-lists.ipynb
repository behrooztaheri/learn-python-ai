A list is an ordered, changeable collection. Positions start at zero and negative positions count from the end.

A **list comprehension** is a short way to build a new list from another one.

```python
nums = [5, 2, 9, 1]
nums.append(7)
nums.remove(2)
nums.sort()
print(nums, nums[-1], nums[1:3])
print(len(nums), sum(nums), max(nums))

squares = [n ** 2 for n in nums]
evens = [n for n in range(10) if n % 2 == 0]
print(squares, evens)
```
