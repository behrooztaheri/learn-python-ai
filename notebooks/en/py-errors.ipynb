# Handling and reading errors

A lesson from the Python and AI School, by Dr. Behrooz Taheri

An error message is not your enemy. Its **last line** names the error and the lines above show where it happened.

Common errors: `SyntaxError`, `NameError`, `TypeError`, `IndexError`, `KeyError`.

Use `try` and `except` for errors you expect, but do not blindly hide every error.

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))

for text in ["12", "abc"]:
    try:
        print(int(text) * 2)
    except ValueError as err:
        print("Bad value:", err)
```
