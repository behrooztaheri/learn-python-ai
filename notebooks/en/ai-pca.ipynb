PCA finds the directions of maximum variance and projects data onto the first few. It helps visualization, compression and reducing collinearity. PCA ignores labels.

Rotate the axis and see which direction maximizes the variance.

## Steps
- Center the data, and usually standardize it.
- Build the covariance matrix.
- Find eigenvectors and eigenvalues.
- Sort the eigenvalues and keep the first k vectors.
- Project the data onto them.

## A 2-D example
If the covariance is $\begin{bmatrix}2&1\\1&2\end{bmatrix}$, the eigenvalues are 3 and 1. So the first component holds $3/(3+1)=75\%$ of the variance.



```python
import numpy as np

values, vectors = np.linalg.eigh(np.array([[2.0, 1.0], [1.0, 2.0]]))
order = values.argsort()[::-1]
print("eigenvalues:", values[order])
print("first direction:", vectors[:, order[0]].round(3))
print("variance ratio:", (values[order] / values.sum()).round(3))
```

```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X = StandardScaler().fit_transform(load_digits().data)
p = PCA().fit(X)
cum = np.cumsum(p.explained_variance_ratio_)
print("Components for 95%:", int(np.searchsorted(cum, 0.95) + 1), "of", X.shape[1])
print("First 3 ratios:", p.explained_variance_ratio_[:3].round(3))
```
