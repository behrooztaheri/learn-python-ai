PCA finds the directions of maximum variance and projects data onto the first few. It helps visualization, compression and reducing collinearity. PCA ignores labels.

Rotate the axis and see which direction maximizes the variance.

```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X = StandardScaler().fit_transform(load_digits().data)
p = PCA().fit(X)
cum = np.cumsum(p.explained_variance_ratio_)
print("Components for 95%:", int(np.searchsorted(cum, 0.95) + 1), "of", X.shape[1])
print("First 3 ratios:", p.explained_variance_ratio_[:3].round(3))
```
