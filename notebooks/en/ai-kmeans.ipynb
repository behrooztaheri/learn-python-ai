Two steps repeat: each point joins its nearest center, and each center moves to the mean of its points.

$$J=\sum_{j=1}^{k}\sum_{x\in C_j}\|x-\mu_j\|^2$$

## Limitations
| Problem | Remedy |
|---|---|
| Sensitive to initialization | `k-means++` and `n_init` |
| Non-spherical clusters | DBSCAN |
| Sensitive to outliers | K-Medoids |
| k must be chosen | Elbow and silhouette |

```python
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

X = StandardScaler().fit_transform(load_iris().data)
for k in range(2, 7):
    m = KMeans(k, n_init=10, random_state=0).fit(X)
    print(k, "WCSS", round(m.inertia_, 1), "silhouette", round(silhouette_score(X, m.labels_), 3))
```
