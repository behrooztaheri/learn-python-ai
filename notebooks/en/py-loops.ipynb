`for` goes through the items of a collection, and `while` repeats as long as a condition is true.

- `range(n)` gives 0 up to n minus 1.
- `enumerate` gives the position and the value together.
- `break` leaves the loop and `continue` jumps to the next round.

> If a `while` loop never ends, press Stop.

## Common patterns
| Pattern | Code | Use |
|---|---|---|
| Counting | `for i in range(n)` | Repeat n times |
| With position | `enumerate(items)` | Position and value |
| Two lists | `zip(a, b)` | Walk in parallel |
| Reverse | `reversed(items)` | From the end |

## Tracing a loop
To understand a loop, write the values round by round. For `total += n` over 1 to 4:

| Round | n | total after |
|---|---|---|
| 1 | 1 | 1 |
| 2 | 2 | 3 |
| 3 | 3 | 6 |
| 4 | 4 | 10 |



```python
names = ["Ali", "Sara"]
scores = [17, 19]
for n, s in zip(names, scores):
    print(n, s)

for row in range(1, 4):
    print(" ".join(str(row * col) for col in range(1, 4)))

for n in range(2, 12):
    if all(n % d for d in range(2, n)):
        print(n, end=" ")
```

```python
total = 0
for n in range(1, 6):
    total += n
print("Sum:", total)

for i, fruit in enumerate(["apple", "kiwi", "plum"]):
    print(i, fruit)

n = 100
steps = 0
while n > 1:
    n = n // 2
    steps += 1
print("Halvings:", steps)
```
