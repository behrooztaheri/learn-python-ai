An SVM finds the boundary with the largest margin, and only the **support vectors** determine it. The kernel trick makes curved boundaries. `C` sets error tolerance and `gamma` sets each sample's reach.

> [The maximum margin and a non-linear RBF boundary.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/svm.png)

*The maximum margin and a non-linear RBF boundary.*

$$\min_{w,b}\ \frac{1}{2}\|w\|^2+C\sum_i\max(0,1-y_i(w^Tx_i+b))$$

The margin width is $2/\|w\|$, so shrinking the weights widens the margin.

| Kernel | Formula | Use |
|---|---|---|
| Linear | $x^Tz$ | Many features, text |
| Polynomial | $(x^Tz+c)^d$ | Feature interactions |
| RBF | $e^{-\gamma\Vert x-z\Vert ^2}$ | A good default |

| Parameter | When large |
|---|---|
| C | Strict margin, overfitting risk |
| gamma | Local influence, overfitting risk |

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, stratify=y, random_state=0)
grid = GridSearchCV(make_pipeline(StandardScaler(), SVC()), {"svc__C": [1, 10], "svc__gamma": ["scale", 0.01]}, cv=3).fit(X_tr, y_tr)
print(grid.best_params_)
print("Test accuracy:", round(grid.score(X_te, y_te), 3))
```
