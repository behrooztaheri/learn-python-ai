Clustering has no true labels, so quality is measured with internal metrics.

| Metric | Formula | Better |
|---|---|---|
| WCSS | $\sum_j\sum_{x\in C_j}\Vert x-\mu_j\Vert ^2$ | Lower |
| Silhouette | $s(i)=\frac{b(i)-a(i)}{\max(a(i),b(i))}$ | Near 1 |
| Davies-Bouldin | $\frac{1}{k}\sum_i\max_{j\ne i}\frac{S_i+S_j}{M_{ij}}$ | Lower |

## Silhouette with one number
$a$ is the mean distance to the point's own cluster and $b$ is the mean distance to the nearest other cluster. If $a=2$ and $b=5$:

$$s=\frac{5-2}{5}=0.6$$

| Value of s | Meaning |
|---|---|
| Near 1 | The point is in the right cluster |
| Near 0 | Near the border of two clusters |
| Negative | Probably in the wrong cluster |

> [The elbow method, mean silhouette, and silhouette per sample on Iris.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/cluster_eval.png)

*The elbow method, mean silhouette, and silhouette per sample on Iris.*



```python
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.metrics import adjusted_rand_score, davies_bouldin_score, silhouette_score
from sklearn.preprocessing import StandardScaler

d = load_iris()
X = StandardScaler().fit_transform(d.data)
for k in [2, 3, 4]:
    l = KMeans(k, n_init=10, random_state=0).fit_predict(X)
    print(k, round(silhouette_score(X, l), 3), round(davies_bouldin_score(X, l), 3), round(adjusted_rand_score(d.target, l), 3))
```



> When true labels are available for checking, use external metrics such as the Adjusted Rand Index.
