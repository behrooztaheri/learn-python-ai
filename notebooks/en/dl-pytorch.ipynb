PyTorch is the main deep learning library. Derivatives are computed automatically by **autograd**.

The training loop has four key lines: `zero_grad`, compute the loss, `backward` and `step`.

> PyTorch cannot run in the browser. Use the Colab button or install it with `pip install torch`.

```python
import torch, torch.nn as nn
from sklearn.datasets import load_digits

X, y = load_digits(return_X_y=True)
X = torch.tensor(X / 16, dtype=torch.float32)
y = torch.tensor(y)

model = nn.Sequential(nn.Linear(64, 128), nn.ReLU(), nn.Dropout(0.2), nn.Linear(128, 10))
loss_fn = nn.CrossEntropyLoss()
opt = torch.optim.Adam(model.parameters(), lr=1e-3)

for epoch in range(60):
    model.train()
    opt.zero_grad()
    loss = loss_fn(model(X[:1400]), y[:1400])
    loss.backward()
    opt.step()

model.eval()
with torch.no_grad():
    acc = (model(X[1400:]).argmax(1) == y[1400:]).float().mean()
print(f"Test accuracy: {acc:.3f}")
```
