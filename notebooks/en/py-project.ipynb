**Exercise:** compute the mean of each subject and print the subject with the lowest mean.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
n = 60
df = pd.DataFrame({
    "math": rng.normal(14, 3, n).clip(0, 20).round(1),
    "physics": rng.normal(13, 3.5, n).clip(0, 20).round(1),
    "study_hours": rng.uniform(0, 12, n).round(1),
})
df.loc[rng.choice(n, 4, replace=False), "physics"] = np.nan
df["physics"] = df["physics"].fillna(df["physics"].median())

def status(avg, passing=10):
    return "pass" if avg >= passing else "fail"

df["average"] = df[["math", "physics"]].mean(axis=1).round(2)
df["status"] = df["average"].apply(status)
print(df["status"].value_counts())
print("Correlation of hours and average:", round(df.study_hours.corr(df.average), 3))

plt.hist(df["average"], bins=12, color="#1f6f8b")
plt.title("Average grades")
plt.xlabel("Average")
plt.ylabel("Students")
plt.show()
```
