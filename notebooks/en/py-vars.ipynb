A variable is a name that refers to a value. Python works out the type from the value.

- `int` whole number
- `float` decimal number
- `str` text
- `bool` true or false

A variable name cannot begin with a digit, and names are case-sensitive.

## Naming rules
| Valid | Invalid | Why |
|---|---|---|
| `total_price` | `total price` | Spaces are not allowed |
| `score2` | `2score` | Cannot begin with a digit |
| `_count` | `class` | A reserved Python word |

The usual Python convention is lowercase with underscores (`snake_case`).

## A variable is a label
In Python a variable is not a box. It is a label pointing to an object. The statement below does not copy anything; two names point to one list.



```python
a = [1, 2]
b = a
b.append(3)
print(a, a is b)

c = a.copy()
c.append(4)
print(a, c, a is c)
```



## Conversion table
| Input | Conversion | Result |
|---|---|---|
| `"3.5"` | `float` | `3.5` |
| `3.9` | `int` | `3` (the decimal part is dropped) |
| `0` | `bool` | `False` |
| `"abc"` | `int` | `ValueError` |

```python
age = 25
height = 1.78
name = "Sara"
is_student = True

print(type(age), type(height), type(name), type(is_student))

# Converting between types
print(int("42") + 1)
print(str(3.5) + " meters")
```
