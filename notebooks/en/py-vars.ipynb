A variable is a name that refers to a value. Python works out the type from the value.

- `int` whole number
- `float` decimal number
- `str` text
- `bool` true or false

A variable name cannot begin with a digit, and names are case-sensitive.

```python
age = 25
height = 1.78
name = "Sara"
is_student = True

print(type(age), type(height), type(name), type(is_student))

# Converting between types
print(int("42") + 1)
print(str(3.5) + " meters")
```
