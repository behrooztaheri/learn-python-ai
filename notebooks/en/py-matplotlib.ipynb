A good plot shows patterns a table hides. The most useful kinds are line for trends, scatter for relationships, histogram for distributions, and bar for comparisons.

Always give a plot a title and axis labels.

## Which plot?
| Goal | Plot | Function |
|---|---|---|
| Trend over time | Line | `plot` |
| Relation of two variables | Scatter | `scatter` |
| Distribution | Histogram | `hist` |
| Comparing groups | Bar | `bar` |
| Outliers | Box | `boxplot` |
| Matrix intensity | Heat map | `imshow` |

To save: `plt.savefig("fig.png", dpi=200)`

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(1)
x = rng.normal(170, 8, 300)
y = 0.9 * x - 90 + rng.normal(0, 6, 300)

fig, ax = plt.subplots(1, 2, figsize=(9, 3.5))
ax[0].hist(x, bins=25, color="#e8590c")
ax[0].set(title="Height distribution", xlabel="Height (cm)", ylabel="Count")
ax[1].scatter(x, y, s=10, color="#5b34c4")
ax[1].set(title="Height vs weight", xlabel="Height (cm)", ylabel="Weight (kg)")
plt.tight_layout()
plt.show()
```
