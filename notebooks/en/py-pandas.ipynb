Pandas is for tabular data. A `DataFrame` is a table and a `Series` is a column.

- Reading: `read_csv`, `read_excel`
- Knowing the data: `head`, `info`, `describe`
- Selecting: `loc` by label, `iloc` by position
- Missing values: `isna`, `fillna`, `dropna`
- Grouping: `groupby`

## Key operations
| Task | Code |
|---|---|
| Select a column | `df["a"]` |
| Filter | `df[df.a > 3]` |
| New column | `df["c"] = df.a * 2` |
| Sort | `df.sort_values("a")` |
| Group | `df.groupby("g").a.mean()` |
| Merge | `pd.merge(a, b, on="id")` |
| Apply a function | `df.a.apply(f)` |



```python
import pandas as pd

students = pd.DataFrame({"id": [1, 2, 3], "name": ["Ali", "Sara", "Reza"]})
grades = pd.DataFrame({"id": [1, 1, 2, 3], "course": ["AI", "Math", "AI", "AI"], "mark": [18, 15, 20, 12]})

df = pd.merge(students, grades, on="id")
print(df)
print(df.groupby("name")["mark"].agg(["mean", "max", "count"]))
print(df.pivot_table(index="name", columns="course", values="mark"))
```

```python
import pandas as pd

df = pd.DataFrame({
    "city": ["Tehran", "Shiraz", "Tehran", "Tabriz", "Shiraz"],
    "sales": [120, 85, None, 60, 95],
    "year": [2024, 2024, 2025, 2025, 2025],
})
print(df.isna().sum())
df["sales"] = df["sales"].fillna(df["sales"].mean())
print(df.head(3))
print(df[df.sales > 90])
print(df.groupby("city")["sales"].mean().round(1))
print(df.describe().round(1))
```
