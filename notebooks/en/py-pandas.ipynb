Pandas is for tabular data. A `DataFrame` is a table and a `Series` is a column.

- Reading: `read_csv`, `read_excel`
- Knowing the data: `head`, `info`, `describe`
- Selecting: `loc` by label, `iloc` by position
- Missing values: `isna`, `fillna`, `dropna`
- Grouping: `groupby`

```python
import pandas as pd

df = pd.DataFrame({
    "city": ["Tehran", "Shiraz", "Tehran", "Tabriz", "Shiraz"],
    "sales": [120, 85, None, 60, 95],
    "year": [2024, 2024, 2025, 2025, 2025],
})
print(df.isna().sum())
df["sales"] = df["sales"].fillna(df["sales"].mean())
print(df.head(3))
print(df[df.sales > 90])
print(df.groupby("city")["sales"].mean().round(1))
print(df.describe().round(1))
```
