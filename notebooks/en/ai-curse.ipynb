# The curse of dimensionality

A lesson from the Python and AI School, by Dr. Behrooz Taheri

Richard Bellman introduced the term **curse of dimensionality** in 1957 while working on dynamic programming. As features increase, the volume of the space grows exponentially and data becomes sparse.

## Signs
- Few samples per feature.
- Accuracy drops as features are added.
- The nearest and farthest points become almost equally far.
- Distance-based models degrade.

## Remedies
Feature selection, feature extraction such as PCA, regularization, more data, and domain knowledge.

```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_wine(return_X_y=True)
rng = np.random.default_rng(0)
for extra in [0, 20, 100, 500]:
    Z = np.hstack([StandardScaler().fit_transform(X), rng.normal(size=(len(X), extra))])
    acc = cross_val_score(KNeighborsClassifier(5), Z, y, cv=5).mean()
    print(f"{extra:4d} useless features -> accuracy {acc:.3f}")
```
