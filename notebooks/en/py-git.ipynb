# Git and GitHub at a glance

A lesson from the Python and AI School, by Dr. Behrooz Taheri

Git stores versions of your code, and GitHub adds online hosting and collaboration.

- `git add` stages changes.
- `git commit` creates a save point.
- `git push` sends it to GitHub.
- `git pull` brings changes from GitHub.

> Never put passwords or huge files in a repository; list them in `.gitignore`.

```bash
git init
git add .
git commit -m "First version"
git branch -M main
git remote add origin https://github.com/USERNAME/REPOSITORY.git
git push -u origin main
```
