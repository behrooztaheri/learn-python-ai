- **Hierarchical:** merges clusters step by step and shows the result as a dendrogram.
- **DBSCAN:** treats dense regions as clusters, finds arbitrary shapes and isolates noise. `eps` and `min_samples` are its main parameters.

> [On moon-shaped data, only DBSCAN finds the true shape.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ch5_algos.png)

*On moon-shaped data, only DBSCAN finds the true shape.*

## Linkage in hierarchical clustering
| Linkage | Distance between clusters |
|---|---|
| Single | The two nearest points |
| Complete | The two farthest points |
| Average | The mean distance |
| Ward | The smallest increase in variance |

## Point types in DBSCAN
| Type | Definition |
|---|---|
| Core | At least `min_samples` points within `eps` |
| Border | Near a core point, not core itself |
| Noise | Neither core nor border |

To choose `eps`, sort and plot the distance to the k-th neighbor and take the bend.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering, DBSCAN, KMeans
from sklearn.datasets import make_moons

X, _ = make_moons(300, noise=0.06, random_state=0)
models = [("KMeans", KMeans(2, n_init=10, random_state=0)), ("Ward", AgglomerativeClustering(2)), ("DBSCAN", DBSCAN(eps=0.2, min_samples=5))]
fig, ax = plt.subplots(1, 3, figsize=(10, 3))
for a, (name, m) in zip(ax, models):
    a.scatter(X[:, 0], X[:, 1], c=m.fit_predict(X), s=12, cmap="viridis")
    a.set_title(name)
plt.show()
```
