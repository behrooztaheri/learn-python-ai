In a **Nash equilibrium**, no player gains by changing strategy alone. John Nash proved that every finite game has at least one equilibrium (possibly mixed).

## Finding pure equilibria
For each column mark player 1's best response, and for each row mark player 2's. A cell with both marks is a Nash equilibrium.

| Game | Pure equilibria | Note |
|---|---|---|
| Prisoner's dilemma | (Confess, Confess) | One, inefficient |
| Battle of the Sexes | Two | A coordination problem |
| Stag Hunt | Two | Safe versus rewarding |
| Matching Pennies | None | Only mixed |



```python
import numpy as np

def pure_nash(A, B):
    """Cells where both players are playing a best response."""
    out = []
    for i in range(A.shape[0]):
        for j in range(A.shape[1]):
            if A[i, j] == A[:, j].max() and B[i, j] == B[i, :].max():
                out.append((i, j))
    return out

prisoners_A = np.array([[-1, -3], [0, -2]])
print("Prisoner's dilemma:", pure_nash(prisoners_A, prisoners_A.T))
sexes_A = np.array([[2, 0], [0, 1]])
sexes_B = np.array([[1, 0], [0, 2]])
print("Battle of the sexes:", pure_nash(sexes_A, sexes_B))
```
