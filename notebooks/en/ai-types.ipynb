# Kinds of learning

A lesson from the Python and AI School, by Dr. Behrooz Taheri

- **Supervised:** data has labels. A numeric output means **regression**, a category means **classification**.
- **Unsupervised:** no labels; the model finds structure, as in clustering and dimensionality reduction.
- **Semi-supervised:** a few labels alongside many unlabeled samples.
- **Reinforcement:** an agent learns by trial, error and reward.

## Supervised and unsupervised, side by side
| Aspect | Supervised | Unsupervised |
|---|---|---|
| Output labels | Yes | No |
| Goal | Predict for new data | Discover structure |
| Example methods | Regression, SVM, trees | K-Means, DBSCAN, PCA |
| Evaluation | Compare to true answers | Internal metrics |
| Example use | Spam detection | Customer segmentation |

> [In supervised learning, the model learns boundaries between labeled classes.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ch5_sup.png)

*In supervised learning, the model learns boundaries between labeled classes.*

> [In unsupervised learning, groups are found without labels.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ch5_unsup.png)

*In unsupervised learning, groups are found without labels.*

## Classification or regression?
| Question | Classification | Regression |
|---|---|---|
| Output | A category | A number |
| Example | Healthy or ill | Price or temperature |
| Metrics | Accuracy, F1 | MAE, RMSE, R² |

> [Classification builds a boundary and regression fits a numeric function.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ch5_cls_reg.png)

*Classification builds a boundary and regression fits a numeric function.*

> To choose the problem type, ask: what does my final output look like, a label or a number?
