`with open` opens a file and closes it automatically. Mode `w` writes, `a` appends and `r` reads. Always set `encoding="utf-8"` for text.

The `pathlib` module handles paths correctly on every operating system, and `json` stores structured data.

```python
from pathlib import Path
import json

path = Path("notes.txt")
with open(path, "w", encoding="utf-8") as f:
    f.write("first line\nsecond line\n")

for line in path.read_text(encoding="utf-8").splitlines():
    print(line.upper())

data = {"course": "Python", "lessons": 20}
Path("data.json").write_text(json.dumps(data))
print(json.loads(Path("data.json").read_text())["lessons"])
```
