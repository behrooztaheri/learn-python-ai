Several layers of neurons form an **MLP**. Training has four steps:

- Forward pass to compute the output
- Compute the loss
- Backpropagation to compute gradients with the chain rule
- Update the weights

$$\frac{\partial J}{\partial w}=\frac{\partial J}{\partial \hat y}\cdot\frac{\partial \hat y}{\partial h}\cdot\frac{\partial h}{\partial w}$$

In the lab, train a real network and watch its decision boundary form.

## One full step with numbers
A sigmoid neuron with $x=1$, $y=1$, $w=0.5$, $b=0$ and learning rate 1:

| Stage | Calculation | Result |
|---|---|---|
| Score | $z=wx+b$ | 0.5 |
| Output | $\hat y=1/(1+e^{-z})$ | 0.6225 |
| Loss | $L=-\ln\hat y$ | 0.474 |
| Gradient | $\partial L/\partial w=(\hat y-y)x$ | −0.3775 |
| Update | $w\leftarrow w-\eta g$ | 0.8775 |



```python
import numpy as np

x, y, w, b, lr = 1.0, 1.0, 0.5, 0.0, 1.0
z = w * x + b
p = 1 / (1 + np.exp(-z))
loss = -np.log(p)
grad = (p - y) * x
print(round(p, 4), round(loss, 3), round(grad, 4), round(w - lr * grad, 4))
```



## Counting parameters
A network with 784 inputs, 128 hidden units and 10 outputs has $(784\times128+128)+(128\times10+10)=101{,}770$ parameters.

```python
import numpy as np

# XOR: a problem one neuron cannot solve
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], float)
y = np.array([[0], [1], [1], [0]], float)

rng = np.random.default_rng(1)
W1, b1 = rng.normal(size=(2, 4)), np.zeros((1, 4))
W2, b2 = rng.normal(size=(4, 1)), np.zeros((1, 1))
sig = lambda z: 1 / (1 + np.exp(-z))

for epoch in range(5001):
    h = np.tanh(X @ W1 + b1)                # forward
    p = sig(h @ W2 + b2)
    loss = -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))
    g2 = p - y                              # backward
    dW2, db2 = h.T @ g2 / 4, g2.mean(0, keepdims=True)
    g1 = (g2 @ W2.T) * (1 - h ** 2)
    dW1, db1 = X.T @ g1 / 4, g1.mean(0, keepdims=True)
    W1 -= 0.5 * dW1; b1 -= 0.5 * db1        # update
    W2 -= 0.5 * dW2; b2 -= 0.5 * db2
    if epoch % 1000 == 0:
        print(epoch, round(float(loss), 4))

print(p.round(3).ravel())
```
