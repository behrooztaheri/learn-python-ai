- **SVR:** ignores errors inside a tube of width $\varepsilon$.
- **Regression tree:** each leaf predicts the mean of its samples.
- **Random forest:** the average of diverse trees.
- **KNN regression:** the mean of k nearest neighbors.

Always compare with a simple linear model.

## Model comparison
| Model | Scaling | Interpretability | Big data | Extrapolation |
|---|---|---|---|---|
| Linear | Suggested | High | Excellent | Linear |
| SVR | Required | Low | Weak | Limited |
| Tree | Not needed | High | Good | None |
| Forest | Not needed | Medium | Good | None |
| KNN | Required | Low | Weak | None |

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
X, y = load_diabetes(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0)

import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor

models = {
    "Linear": LinearRegression(),
    "SVR": make_pipeline(StandardScaler(), SVR(C=100, gamma=0.05)),
    "Tree": DecisionTreeRegressor(max_depth=3, random_state=0),
    "Forest": RandomForestRegressor(200, min_samples_leaf=3, random_state=0),
    "KNN": make_pipeline(StandardScaler(), KNeighborsRegressor(15)),
}
for name, m in models.items():
    print(f"{name:7s} RMSE = {np.sqrt(mean_squared_error(y_te, m.fit(X_tr, y_tr).predict(X_te))):.2f}")
```
