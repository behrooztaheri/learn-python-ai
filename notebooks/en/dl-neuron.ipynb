A neuron adds up weighted inputs and passes the result through an activation function.

$$\hat y=f\Big(\sum_i w_ix_i+b\Big)$$

Without a non-linear function, several layers collapse into one linear map. **ReLU** is the usual choice for hidden layers.

> [Common activation functions.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/activations.png)

*Common activation functions.*

| Function | Range | Where | Note |
|---|---|---|---|
| ReLU | $[0,\infty)$ | Hidden layers | Fast; dying neurons risk |
| Leaky ReLU | $(-\infty,\infty)$ | Hidden layers | Small slope for negatives |
| Sigmoid | $(0,1)$ | Binary output | Vanishing gradients |
| Tanh | $(-1,1)$ | RNNs | Zero-centered |
| Softmax | Sums to 1 | Multi-class output | Class probabilities |

## Choosing output and loss
| Problem | Output | Loss |
|---|---|---|
| Regression | Linear | MSE |
| Binary | Sigmoid | Binary cross-entropy |
| Multi-class | Softmax | Cross-entropy |

```python
import numpy as np

def sigmoid(z): return 1 / (1 + np.exp(-z))
def relu(z): return np.maximum(z, 0)

x = np.array([0.5, -1.2, 3.0])
w = np.array([0.8, 0.1, -0.4])
z = w @ x + 0.2
print("score:", round(z, 3))
print("sigmoid:", round(sigmoid(z), 3), "relu:", relu(z))
```
