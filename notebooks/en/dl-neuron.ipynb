A neuron adds up weighted inputs and passes the result through an activation function.

$$\hat y=f\Big(\sum_i w_ix_i+b\Big)$$

Without a non-linear function, several layers collapse into one linear map. **ReLU** is the usual choice for hidden layers.

```python
import numpy as np

def sigmoid(z): return 1 / (1 + np.exp(-z))
def relu(z): return np.maximum(z, 0)

x = np.array([0.5, -1.2, 3.0])
w = np.array([0.8, 0.1, -0.4])
z = w @ x + 0.2
print("score:", round(z, 3))
print("sigmoid:", round(sigmoid(z), 3), "relu:", relu(z))
```
