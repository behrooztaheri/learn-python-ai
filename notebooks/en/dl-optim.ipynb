- **SGD:** gradients on small batches
- **Momentum:** averages gradients to reduce oscillation
- **Adam:** adaptive steps and a good starting point

Tools against overfitting: **Dropout**, **Weight Decay**, **Early Stopping**, **Batch Normalization** and data augmentation.

> [SGD, Momentum and Adam in a narrow valley.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/optim.gif)

*SGD, Momentum and Adam in a narrow valley.*

| Optimizer | Idea | Common default |
|---|---|---|
| SGD | Mini-batch gradient | Rate 0.1 to 0.001 |
| Momentum | Moves with inertia | Factor 0.9 |
| RMSProp | Scales by mean squared gradient | Rate 0.001 |
| Adam | Combines both | Rate 0.001 |

$$v_{t+1}=\beta v_t-\eta\nabla J,\quad \theta_{t+1}=\theta_t+v_{t+1}$$

## Guide to controlling overfitting
| Tool | How it helps |
|---|---|
| Dropout | Randomly silences neurons |
| Weight decay | Penalizes large weights |
| Early stopping | Stops at the best validation |
| Batch norm | Stabilizes training |
| Augmentation | More varied data |

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, stratify=y, random_state=0)
for solver in ["sgd", "adam"]:
    m = make_pipeline(StandardScaler(), MLPClassifier((64,), solver=solver, max_iter=60, random_state=0))
    print(solver, round(m.fit(X_tr, y_tr).score(X_te, y_te), 3))
```
