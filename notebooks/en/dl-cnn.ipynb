A small kernel slides over an image and produces a **feature map**. **Weight sharing** keeps parameters few. Pooling reduces size, so deeper layers see larger patterns.

Change the kernel and paint on the pixel grid.

```python
import torch, torch.nn as nn

model = nn.Sequential(
    nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
    nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
    nn.Flatten(), nn.Linear(32 * 7 * 7, 10),
)
x = torch.zeros(4, 1, 28, 28)
print(model(x).shape, sum(p.numel() for p in model.parameters()))
```
