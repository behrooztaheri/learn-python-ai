A function is a named block of code that takes inputs and gives back a result with `return`. Functions avoid repetition and make code easier to read.

Inputs may have default values. `lambda` makes a short unnamed function.

## Parts of a function
| Part | Meaning |
|---|---|
| Parameter | The input's name in the definition |
| Argument | The actual value when calling |
| Default value | Used if no argument is given |
| `*args` | Any number of positional inputs |
| `**kwargs` | Any number of named inputs |

## Variable scope
A variable created inside a function does not exist outside it. Take data as parameters and give it back with `return`.

> Never use a list as a default. Write `def f(a=None)` instead of `def f(a=[])`.



```python
def stats(*values, precision=2):
    mean = sum(values) / len(values)
    return round(mean, precision), max(values)

avg, top = stats(3, 7, 8, precision=1)
print(avg, top)

def describe(**info):
    for k, v in info.items():
        print(k, "->", v)

describe(name="Ali", city="Rasht")
```

```python
def bmi(weight, height=1.70):
    """Return body mass index."""
    return weight / height ** 2

print(round(bmi(70), 1))
print(round(bmi(70, height=1.85), 1))

people = [("Ali", 20), ("Sara", 18), ("Reza", 25)]
print(sorted(people, key=lambda p: p[1]))
```
