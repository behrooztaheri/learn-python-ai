Gradient descent finds the minimum of a cost function with small steps against the gradient.

$$\theta_{t+1}=\theta_t-\eta\nabla J(\theta_t)$$

Change the learning rate $\eta$: too small is slow, too large diverges.

```python
import numpy as np

rng = np.random.default_rng(0)
x = rng.uniform(-1, 1, 100)
y = 3 * x + 2 + rng.normal(0, 0.2, 100)

w, b, lr = 0.0, 0.0, 0.1
for step in range(201):
    e = (w * x + b) - y
    w -= lr * 2 * np.mean(e * x)
    b -= lr * 2 * np.mean(e)
    if step % 50 == 0:
        print(step, round(np.mean(e ** 2), 4), round(w, 3), round(b, 3))
```
