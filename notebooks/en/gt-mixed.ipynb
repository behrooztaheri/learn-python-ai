Sometimes a player must randomize to stay unpredictable. In a mixed equilibrium, each player's probabilities make **the opponent indifferent** between options.

## Example: Battle of the Sexes
| | Column A | Column B |
|---|---|---|
| **Row A** | (2, 1) | (0, 0) |
| **Row B** | (0, 0) | (1, 2) |

Player 1 plays A with probability p. Player 2 must be indifferent:

$$1\cdot p=2(1-p)\Rightarrow p=\tfrac{2}{3}$$

Likewise player 2's $q$ makes player 1 indifferent: $2q=1-q$, so $q=\tfrac{1}{3}$.



```python
from fractions import Fraction as F

a, b, c, d = F(1), F(0), F(0), F(2)   # player 2's payoffs for rows A,B and columns A,B
p = (d - c) / (a - b - c + d)
A, C = F(2), F(1)
q = C / (A + C)
print("p =", p, "q =", q)
print("Expected payoff of player 1:", A * q)
```
