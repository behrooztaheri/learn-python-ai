In **self-attention**, every token looks at all tokens.

$$Attention(Q,K,V)=softmax\Big(\frac{QK^T}{\sqrt{d_k}}\Big)V$$

Large language models are built on this architecture: text becomes tokens and the model predicts the next token probabilistically.

> A language model may produce a wrong answer confidently. Always verify important results.

## Parts of a Transformer
| Part | Role |
|---|---|
| Embedding | Token to vector |
| Positional encoding | Adds order information |
| Multi-head attention | Several kinds of attention in parallel |
| Feed forward | Processes each token |
| Residual and layer norm | Stable deep training |

In attention, a **query** asks, a **key** advertises and a **value** supplies content. The more similar a query and key, the higher the weight.

```python
import numpy as np

# Scaled dot-product attention on three tokens
Q = K = V = np.array([[1.0, 0.0], [0.9, 0.1], [0.0, 1.0]])
scores = Q @ K.T / np.sqrt(K.shape[1])
exp = np.exp(scores - scores.max(axis=1, keepdims=True))
weights = exp / exp.sum(axis=1, keepdims=True)
print(weights.round(3))
print((weights @ V).round(3))
```
