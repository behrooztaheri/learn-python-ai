NumPy provides multi-dimensional arrays and fast computing. Operations work on whole arrays without loops (**vectorization**), and **broadcasting** combines compatible arrays of different sizes.

- `shape` the array's dimensions
- `axis=0` along rows (per column), `axis=1` along columns (per row)
- `@` matrix multiplication

## Creating arrays
| Function | Result |
|---|---|
| `np.zeros((2,3))` | Zero matrix |
| `np.ones(4)` | Vector of ones |
| `np.arange(0,10,2)` | With a set step |
| `np.linspace(0,1,5)` | With a set count |
| `np.eye(3)` | Identity matrix |

## Broadcasting in action
A `(2,3)` array can be added to a `(3,)` array because the last dimensions match and the smaller row is repeated.



```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
print(A + np.array([10, 20, 30]))
print(A[:, 1], A[1, :], A[0, 2])

B = np.array([[2, 1], [1, 3]])
b = np.array([3, 5])
print("solve Bx=b:", np.linalg.solve(B, b))
print("inverse:\n", np.linalg.inv(B).round(2))
```

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.shape, a.mean(), a.mean(axis=0))
print(a * 10)
print(a[a > 2])
print(a @ a.T)

x = np.linspace(0, 1, 5)
print(x, x.reshape(5, 1).shape)
rng = np.random.default_rng(0)
print(rng.normal(size=3).round(3))
```
