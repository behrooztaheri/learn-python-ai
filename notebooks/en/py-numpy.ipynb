NumPy provides multi-dimensional arrays and fast computing. Operations work on whole arrays without loops (**vectorization**), and **broadcasting** combines compatible arrays of different sizes.

- `shape` the array's dimensions
- `axis=0` along rows (per column), `axis=1` along columns (per row)
- `@` matrix multiplication

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.shape, a.mean(), a.mean(axis=0))
print(a * 10)
print(a[a > 2])
print(a @ a.T)

x = np.linspace(0, 1, 5)
print(x, x.reshape(5, 1).shape)
rng = np.random.default_rng(0)
print(rng.normal(size=3).round(3))
```
