In a Sugeno system, the rule consequent is a function:

$$IF\ x\ is\ A\ AND\ y\ is\ B\ THEN\ z=px+qy+r$$

The output is a weighted average and is simpler to compute than Mamdani. That is why it underlies ANFIS.

| Property | Mamdani | Sugeno |
|---|---|---|
| Rule consequent | Fuzzy set | Function |
| Defuzzification | Centroid | Weighted average |
| Interpretability | High | Medium |
| Computation | Heavy | Light |
| Training on data | Difficult | Suitable |

```python
import numpy as np

gauss = lambda x, c, s: np.exp(-(x - c) ** 2 / (2 * s ** 2))

x, y = 3.0, 4.0
w1 = gauss(x, 2, 2) * gauss(y, 3, 2)
w2 = gauss(x, 6, 2) * gauss(y, 7, 2)
f1 = 0.5 * x + 0.2 * y + 1
f2 = -0.3 * x + 0.8 * y + 2
print("output:", round((w1 * f1 + w2 * f2) / (w1 + w2), 3))
```
