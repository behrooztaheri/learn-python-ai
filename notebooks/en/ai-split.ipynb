# Splitting data and validation

A lesson from the Python and AI School, by Dr. Behrooz Taheri

A model must be judged on data it has not seen.

- **Training:** learning the parameters
- **Validation:** choosing models and tuning hyperparameters
- **Test:** the final evaluation, used once

In **k-fold** the data is divided into k parts, and each part serves once as validation. With time series, keep the time order.

> [In each round, one part validates and the others train.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/kfold.png)

*In each round, one part validates and the others train.*

## Which method?
| Method | When |
|---|---|
| `train_test_split` | Lots of data, quick check |
| `KFold` | Moderate data |
| `StratifiedKFold` | Classification, especially imbalanced |
| `GroupKFold` | Several samples per patient or user |
| `TimeSeriesSplit` | Time-ordered data |

## Tuning hyperparameters
`GridSearchCV` tries all combinations and `RandomizedSearchCV` tries a random sample. Random search is usually more economical with many parameters.

> A random split is wrong for time series, because the model gains information from the future.

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_iris(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)

model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=500))
scores = cross_val_score(model, X_tr, y_tr, cv=5)
print("CV accuracy:", scores.round(3), "mean:", scores.mean().round(3))
print("Test accuracy:", round(model.fit(X_tr, y_tr).score(X_te, y_te), 3))
```
