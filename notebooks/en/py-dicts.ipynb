# Dictionaries, tuples and sets

A lesson from the Python and AI School, by Dr. Behrooz Taheri

- A **dictionary** maps each key to a value.
- A **tuple** is like a list but cannot be changed.
- A **set** has no duplicates and is fast for membership tests.

```python
student = {"name": "Ali", "grade": 18}
student["city"] = "Tehran"
for key, value in student.items():
    print(key, "=", value)
print(student.get("age", "unknown"))

point = (3, 4)
x, y = point
print(x + y)

words = ["a", "b", "a", "c", "b"]
print(set(words), len(set(words)))
```
