# Dictionaries, tuples and sets

A lesson from the Python and AI School, by Dr. Behrooz Taheri

- A **dictionary** maps each key to a value.
- A **tuple** is like a list but cannot be changed.
- A **set** has no duplicates and is fast for membership tests.

## Which structure?
| Structure | Ordered | Mutable | Duplicates | Good for |
|---|---|---|---|---|
| list | Yes | Yes | Allowed | Sequences |
| tuple | Yes | No | Allowed | Fixed data |
| set | No | Yes | Not allowed | Uniqueness and membership |
| dict | Insertion order | Yes | Unique keys | Key to value maps |

## Counting pattern


```python
text = "to be or not to be"
counts = {}
for w in text.split():
    counts[w] = counts.get(w, 0) + 1
print(counts)
print(sorted(counts.items(), key=lambda kv: kv[1], reverse=True))

a, b = {1, 2, 3}, {2, 3, 4}
print(a & b, a | b, a - b)
```

```python
student = {"name": "Ali", "grade": 18}
student["city"] = "Tehran"
for key, value in student.items():
    print(key, "=", value)
print(student.get("age", "unknown"))

point = (3, 4)
x, y = point
print(x + y)

words = ["a", "b", "a", "c", "b"]
print(set(words), len(set(words)))
```
