A model is only as good as its data.

- **Missing values:** drop them, or fill with the mean or median.
- **Outliers:** check whether they are recording errors or real events.
- **Categorical variables:** convert to numbers with one-hot encoding.
- **Scaling:** needed for models that use distances or gradients.

$$z=\frac{x-\mu}{\sigma}$$

> **Data leakage:** fit preprocessing on training data only. A `Pipeline` does this automatically and safely.

## Handling missing values
| Method | When | Risk |
|---|---|---|
| Drop rows | Very few and random | Losing data |
| Mean | Roughly symmetric data | Reduces spread |
| Median | Skewed data or outliers | Reduces spread |
| Most frequent | Categorical variables | Bias toward the main class |

## Detecting outliers with IQR
Treat values outside this range as suspicious:

$$[Q_1-1.5\,IQR,\ Q_3+1.5\,IQR],\quad IQR=Q_3-Q_1$$



```python
import numpy as np

x = np.array([12, 13, 14, 15, 15, 16, 17, 58])
q1, q3 = np.percentile(x, [25, 75])
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print("range:", low, high)
print("outliers:", x[(x < low) | (x > high)])
print("mean:", x.mean(), "median:", np.median(x))
```



## Scaling
| Method | Formula | Good for |
|---|---|---|
| Standardization | $(x-\mu)/\sigma$ | Most models |
| Min-Max | $(x-x_{min})/(x_{max}-x_{min})$ | Bounded inputs |
| Robust | $(x-median)/IQR$ | Data with outliers |

## Encoding categories
| Method | When |
|---|---|
| One-Hot | Unordered categories such as city |
| Ordinal | Ordered categories such as low, medium, high |

```python
import pandas as pd
from sklearn.compose import make_column_transformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

df = pd.DataFrame({"age": [25, 32, None, 51], "city": ["A", "B", "A", "C"]})

numeric = make_pipeline(SimpleImputer(strategy="median"), StandardScaler())
prep = make_column_transformer((numeric, ["age"]), (OneHotEncoder(), ["city"]))
print(prep.fit_transform(df).round(2))
```
