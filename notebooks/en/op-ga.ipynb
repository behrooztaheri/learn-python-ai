A genetic algorithm is inspired by evolution.

| Stage | Action |
|---|---|
| Initial population | Several random solutions |
| Evaluation | Fitness of each solution |
| Selection | Better solutions get more chances |
| Crossover | Combine two parents |
| Mutation | A small random change |
| Elitism | Keep the best solution |



```python
import numpy as np

def rastrigin(x):
    x = np.atleast_2d(x)
    return 10 * x.shape[1] + np.sum(x ** 2 - 10 * np.cos(2 * np.pi * x), axis=1)

LOW, HIGH, DIM = -5.12, 5.12, 2
rng = np.random.default_rng(1)

pop = rng.uniform(LOW, HIGH, (60, DIM))
for gen in range(120):
    fit = rastrigin(pop)
    elite = pop[fit.argmin()].copy()
    a, b = rng.integers(0, 60, (2, 60))            # tournament selection
    parents = pop[np.where(fit[a] < fit[b], a, b)]
    alpha = rng.random((60, 1))                    # blend crossover
    children = alpha * parents + (1 - alpha) * parents[::-1]
    mask = rng.random(children.shape) < 0.15       # mutation
    children[mask] += rng.normal(0, 0.4, mask.sum())
    pop = np.clip(children, LOW, HIGH)
    pop[0] = elite                                 # elitism
best = pop[rastrigin(pop).argmin()]
print("best:", best.round(4), "value:", rastrigin(best).round(5))
```



| Parameter | High | Low |
|---|---|---|
| Population size | Better search, slower | Fast, local risk |
| Mutation rate | Much diversity | Getting stuck |
