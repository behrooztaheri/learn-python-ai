Numeric operators: `+`, `-`, `*`, `/` (decimal division), `//` (whole division), `%` (remainder) and `**` (power).

A string is a sequence of characters, indexed from zero. An **f-string** is the easiest way to put values inside text.

Use the lab below to see exactly what the slice `[start:stop:step]` selects.

## Useful string methods
| Method | Purpose | Example | Result |
|---|---|---|---|
| `strip` | Trim spaces | `" a ".strip()` | `"a"` |
| `lower` | Lowercase | `"Ab".lower()` | `"ab"` |
| `split` | Break into a list | `"a,b".split(",")` | `["a","b"]` |
| `join` | Glue a list | `"-".join(["a","b"])` | `"a-b"` |
| `replace` | Replace text | `"cat".replace("c","b")` | `"bat"` |
| `find` | Locate text | `"data".find("t")` | `2` |

## Operator precedence
In order: parentheses, power, multiply and divide, add and subtract. When unsure, add parentheses.

> Decimal numbers are not stored exactly in binary. Compare them with `math.isclose`.



```python
import math

print(0.1 + 0.2 == 0.3, 0.1 + 0.2)
print(math.isclose(0.1 + 0.2, 0.3))
print(2 + 3 * 4, (2 + 3) * 4, 2 ** 3 ** 2)
print("-".join(["a", "b", "c"]), "  hi ".strip())
```

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)

text = "Machine Learning"
print(text[0], text[-1], text[0:7])
print(text.upper(), text.replace("Machine", "Deep"))
print(len(text), text.split())

score = 18.456
print(f"Score: {score:.1f} out of 20")
```
