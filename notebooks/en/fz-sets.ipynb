In a classical set, membership is 0 or 1. Lotfi Zadeh introduced **fuzzy sets** in 1965: membership is a degree between 0 and 1.

A **membership function** gives this degree; triangular, trapezoidal, Gaussian and bell shapes are common.

Operators: AND with min, OR with max, and NOT with $1-\mu$.

> Fuzziness differs from probability: it is a degree of truth, not a chance of occurring.

## Membership formulas
| Name | Formula |
|---|---|
| Triangular | $\max(\min(\frac{x-a}{b-a},\frac{c-x}{c-b}),0)$ |
| Gaussian | $e^{-(x-c)^2/2\sigma^2}$ |
| Bell | $1/(1+\vert \frac{x-c}{a}\vert ^{2b})$ |

> [The AND, OR and NOT operators.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ops.png)

*The AND, OR and NOT operators.*

```python
import numpy as np

def tri(x, a, b, c):
    return np.maximum(np.minimum((x - a) / (b - a), (c - x) / (c - b)), 0)

x = np.array([10, 20, 25, 30])
warm = tri(x, 15, 25, 35)
hot = tri(x, 25, 35, 45)
print("warm", warm.round(2))
print("hot ", hot.round(2))
print("warm AND hot", np.minimum(warm, hot).round(2))
print("warm OR hot ", np.maximum(warm, hot).round(2))
```
