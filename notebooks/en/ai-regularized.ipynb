# Ridge, Lasso and Elastic Net

A lesson from the Python and AI School, by Dr. Behrooz Taheri

Regularization penalizes weights to prevent overfitting.

$$Ridge:\ \sum(y-\hat y)^2+\alpha\sum w_j^2$$

$$Lasso:\ \tfrac{1}{2n}\sum(y-\hat y)^2+\alpha\sum|w_j|$$

- **Ridge** shrinks weights and suits correlated features.
- **Lasso** sets some weights exactly to zero, selecting features.
- **Elastic Net** combines both.

Inputs must be scaled.

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
X, y = load_diabetes(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0)

import numpy as np
from sklearn.linear_model import ElasticNet, Lasso, LinearRegression, Ridge
from sklearn.metrics import mean_squared_error
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

models = {"OLS": LinearRegression(), "Ridge": Ridge(alpha=10), "Lasso": Lasso(alpha=1), "ElasticNet": ElasticNet(alpha=0.5, l1_ratio=0.5)}
for name, reg in models.items():
    m = make_pipeline(StandardScaler(), reg).fit(X_tr, y_tr)
    rmse = np.sqrt(mean_squared_error(y_te, m.predict(X_te)))
    nz = int(np.sum(np.abs(m[-1].coef_) > 1e-8))
    print(f"{name:11s} RMSE={rmse:.2f} non-zero weights={nz}")
```
