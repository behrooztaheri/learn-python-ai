Rules are written in plain words: "If the weather is warm, then the fan speed is high."

- **Fuzzification:** membership degrees of the inputs
- **Rule evaluation:** clipping outputs by rule strength
- **Aggregation:** combining results with max
- **Defuzzification:** back to a number with the centroid

$$z^*=\frac{\int\mu(z)z\,dz}{\int\mu(z)\,dz}$$

## Defuzzification methods
> [Comparing defuzzification methods on one output.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/defuzz.png)

*Comparing defuzzification methods on one output.*

| Method | Idea | Property |
|---|---|---|
| Centroid | Center of gravity | Smooth, common |
| Bisector | Halves the area | Near centroid |
| MOM | Mean of maxima | Fast |
| SOM and LOM | Start or end of maxima | Sensitive |

> [A control surface: a few word rules form a non-linear mapping.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/surface.png)

*A control surface: a few word rules form a non-linear mapping.*

```python
import numpy as np

def tri(x, a, b, c):
    return np.maximum(np.minimum((x - a) / (b - a + 1e-9), (c - x) / (c - b + 1e-9)), 0)

z = np.linspace(0, 100, 1001)
slow, medium, fast = tri(z, -1, 0, 50), tri(z, 0, 50, 100), tri(z, 50, 100, 101)

def fan_speed(t):
    cold, warm, hot = tri(t, 0, 10, 25), tri(t, 15, 25, 35), tri(t, 25, 40, 55)
    agg = np.maximum.reduce([np.minimum(cold, slow), np.minimum(warm, medium), np.minimum(hot, fast)])
    return (z * agg).sum() / agg.sum()

for t in [12.0, 22.0, 30.0, 40.0]:
    print(f"{t} C -> fan {fan_speed(t):.1f}%")
```
