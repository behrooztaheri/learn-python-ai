**Ensemble learning** combines several models.

- **Bagging and random forests:** independent trees on random samples, reducing variance.
- **Boosting:** each model corrects the errors of earlier ones, as in Gradient Boosting, XGBoost and LightGBM.

On tabular data, these are among the strongest options.

| Property | Bagging / forest | Boosting |
|---|---|---|
| Model building | Parallel | Sequential |
| Reduces | Variance | Bias and variance |
| Noise sensitivity | Low | Higher |
| Key parameters | Trees, `max_features` | Learning rate, depth, stages |

In Gradient Boosting, each new tree learns the negative gradient of the loss, which for MSE is the residual.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
X, y = load_breast_cancer(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)

from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier

for name, m in [("Tree", DecisionTreeClassifier(random_state=0)), ("Forest", RandomForestClassifier(200, random_state=0)), ("Boosting", GradientBoostingClassifier(random_state=0))]:
    print(name, round(m.fit(X_tr, y_tr).score(X_te, y_te), 3))
```
