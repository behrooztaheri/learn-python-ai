# انواع یادگیری

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

- **با نظارت:** داده برچسب دارد. اگر خروجی عدد باشد **رگرسیون** و اگر دسته باشد **طبقه‌بندی** است.
- **بدون نظارت:** برچسبی نیست و مدل ساختار داده را پیدا می‌کند؛ مانند خوشه‌بندی و کاهش بعد.
- **نیمه‌نظارتی:** مقدار کمی برچسب کنار داده زیاد بدون برچسب.
- **تقویتی:** عامل با آزمون و خطا و دریافت پاداش یاد می‌گیرد.
