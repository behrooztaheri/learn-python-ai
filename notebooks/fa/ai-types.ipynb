# انواع یادگیری

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

- **با نظارت:** داده برچسب دارد. اگر خروجی عدد باشد **رگرسیون** و اگر دسته باشد **طبقه‌بندی** است.
- **بدون نظارت:** برچسبی نیست و مدل ساختار داده را پیدا می‌کند؛ مانند خوشه‌بندی و کاهش بعد.
- **نیمه‌نظارتی:** مقدار کمی برچسب کنار داده زیاد بدون برچسب.
- **تقویتی:** عامل با آزمون و خطا و دریافت پاداش یاد می‌گیرد.

## با نظارت و بدون نظارت، کنار هم
| معیار | با نظارت | بدون نظارت |
|---|---|---|
| برچسب خروجی | دارد | ندارد |
| هدف | پیش‌بینی برای داده جدید | کشف ساختار داده |
| نمونه روش | رگرسیون، SVM، درخت | K-Means، DBSCAN، PCA |
| ارزیابی | مقایسه با جواب واقعی | معیارهای داخلی |
| نمونه کاربرد | تشخیص اسپم | بخش‌بندی مشتری |

> [در یادگیری با نظارت، مدل مرز بین کلاس‌های برچسب‌دار را یاد می‌گیرد.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ch5_sup.png)

*در یادگیری با نظارت، مدل مرز بین کلاس‌های برچسب‌دار را یاد می‌گیرد.*

> [در یادگیری بدون نظارت، گروه‌ها بدون برچسب پیدا می‌شوند.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ch5_unsup.png)

*در یادگیری بدون نظارت، گروه‌ها بدون برچسب پیدا می‌شوند.*

## طبقه‌بندی یا رگرسیون؟
| پرسش | طبقه‌بندی | رگرسیون |
|---|---|---|
| خروجی | دسته | عدد |
| مثال | سالم یا بیمار | قیمت یا دما |
| معیار | Accuracy، F1 | MAE، RMSE، R² |

> [طبقه‌بندی مرز می‌سازد و رگرسیون یک تابع عددی برازش می‌کند.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ch5_cls_reg.png)

*طبقه‌بندی مرز می‌سازد و رگرسیون یک تابع عددی برازش می‌کند.*
