# حلقه‌ها

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

`for` روی اعضای یک مجموعه می‌گردد و `while` تا زمانی که شرط درست باشد تکرار می‌کند.

- `range(n)` اعداد ۰ تا n منهای ۱ را می‌دهد.
- `enumerate` شماره و مقدار را با هم می‌دهد.
- `break` از حلقه خارج می‌شود و `continue` به دور بعد می‌رود.

> اگر حلقه `while` تمام نشد، دکمه «توقف» را بزن.

## الگوهای پرکاربرد
| الگو | کد | کاربرد |
|---|---|---|
| شمارش | `for i in range(n)` | تکرار n بار |
| با شماره | `enumerate(items)` | شماره و مقدار |
| دو لیست با هم | `zip(a, b)` | پیمایش موازی |
| معکوس | `reversed(items)` | از انتها |

## ردگیری یک حلقه
برای فهمیدن حلقه، مقدارها را دور به دور بنویس. در محاسبه `total += n` برای اعداد ۱ تا ۴:

| دور | n | total بعد از اجرا |
|---|---|---|
| ۱ | 1 | 1 |
| ۲ | 2 | 3 |
| ۳ | 3 | 6 |
| ۴ | 4 | 10 |



</div>

```python
names = ["Ali", "Sara"]
scores = [17, 19]
for n, s in zip(names, scores):
    print(n, s)

for row in range(1, 4):
    print(" ".join(str(row * col) for col in range(1, 4)))

for n in range(2, 12):
    if all(n % d for d in range(2, n)):
        print(n, end=" ")
```

## آزمایشگاه حلقه
[[lab:loop]]

</div>

```python
total = 0
for n in range(1, 6):
    total += n
print("Sum:", total)

for i, fruit in enumerate(["apple", "kiwi", "plum"]):
    print(i, fruit)

n = 100
steps = 0
while n > 1:
    n = n // 2
    steps += 1
print("Halvings:", steps)
```
