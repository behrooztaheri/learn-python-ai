# داده و پیش‌پردازش

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

کیفیت مدل به کیفیت داده وابسته است.

- **مقدار گمشده:** حذف، یا جایگزینی با میانگین یا میانه.
- **داده پرت:** بررسی کن خطای ثبت است یا پدیده واقعی.
- **متغیر دسته‌ای:** با One-Hot به عدد تبدیل می‌شود.
- **مقیاس‌بندی:** برای مدل‌های حساس به فاصله و گرادیان لازم است.

$$z=\frac{x-\mu}{\sigma}$$

> **نشت داده:** پیش‌پردازش را فقط روی داده آموزش fit کن. `Pipeline` این کار را خودکار و امن انجام می‌دهد.

## روش برخورد با مقدار گمشده
| روش | چه زمانی | ریسک |
|---|---|---|
| حذف سطر | گمشده بسیار کم و تصادفی | از دست رفتن داده |
| میانگین | داده تقریباً متقارن | کاهش پراکندگی |
| میانه | داده چوله یا دارای پرت | کاهش پراکندگی |
| پرتکرارترین مقدار | متغیر دسته‌ای | سوگیری به کلاس غالب |

## تشخیص داده پرت با IQR
مقدار خارج از بازه زیر را مشکوک بدان:

$$[Q_1-1.5\,IQR,\ Q_3+1.5\,IQR],\quad IQR=Q_3-Q_1$$



</div>

```python
import numpy as np

x = np.array([12, 13, 14, 15, 15, 16, 17, 58])
q1, q3 = np.percentile(x, [25, 75])
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print("range:", low, high)
print("outliers:", x[(x < low) | (x > high)])
print("mean:", x.mean(), "median:", np.median(x))
```

<div dir="rtl" align="right">



## مقیاس‌بندی
| روش | فرمول | مناسب برای |
|---|---|---|
| استانداردسازی | $(x-\mu)/\sigma$ | بیشتر مدل‌ها |
| Min-Max | $(x-x_{min})/(x_{max}-x_{min})$ | ورودی با بازه محدود |
| Robust | $(x-median)/IQR$ | داده دارای پرت |

## کدگذاری متغیر دسته‌ای
| روش | چه زمانی |
|---|---|
| One-Hot | دسته‌های بدون ترتیب، مثل شهر |
| Ordinal | دسته‌های دارای ترتیب، مثل کم، متوسط، زیاد |

</div>

```python
import pandas as pd
from sklearn.compose import make_column_transformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

df = pd.DataFrame({"age": [25, 32, None, 51], "city": ["A", "B", "A", "C"]})

numeric = make_pipeline(SimpleImputer(strategy="median"), StandardScaler())
prep = make_column_transformer((numeric, ["age"]), (OneHotEncoder(), ["city"]))
print(prep.fit_transform(df).round(2))
```
