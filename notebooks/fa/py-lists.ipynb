# لیست

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

لیست مجموعه‌ای مرتب و قابل تغییر است. شماره‌گذاری از صفر است و شماره منفی از انتها می‌شمارد.

**List comprehension** راهی کوتاه برای ساختن لیست جدید از روی لیست دیگر است.

## متدهای لیست
| متد | کار | لیست را تغییر می‌دهد؟ |
|---|---|---|
| `append(x)` | افزودن به انتها | بله |
| `insert(i, x)` | افزودن در محل i | بله |
| `pop(i)` | حذف و برگرداندن | بله |
| `sort()` | مرتب‌سازی | بله |
| `sorted(lst)` | لیست مرتب جدید | خیر |
| `count(x)` | شمارش | خیر |

> خطای رایج: تغییر لیست هنگام پیمایش همان لیست. روی کپی پیمایش کن یا لیست جدید بساز.



</div>

```python
nums = [1, 2, 2, 3, 2]

result = [n for n in nums if n != 2]
print(result)

matrix = [[1, 2, 3], [4, 5, 6]]
print([row[1] for row in matrix])
print([x for row in matrix for x in row])
```

```python
nums = [5, 2, 9, 1]
nums.append(7)
nums.remove(2)
nums.sort()
print(nums, nums[-1], nums[1:3])
print(len(nums), sum(nums), max(nums))

squares = [n ** 2 for n in nums]
evens = [n for n in range(10) if n % 2 == 0]
print(squares, evens)
```
