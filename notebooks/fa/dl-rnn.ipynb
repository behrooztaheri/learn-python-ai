شبکه بازگشتی با **حالت پنهان** دنباله را گام‌به‌گام می‌خواند. RNN ساده در دنباله بلند با محو گرادیان مشکل دارد. **LSTM** و **GRU** با گیت‌ها وابستگی‌های بلندتر را یاد می‌گیرند.

$$h_t=\tanh(W_xx_t+W_hh_{t-1}+b)$$

</div>

```python
import torch, torch.nn as nn

class Forecaster(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(1, 32, batch_first=True)
        self.out = nn.Linear(32, 1)
    def forward(self, x):
        s, _ = self.lstm(x)
        return self.out(s[:, -1])

t = torch.linspace(0, 60, 1200)
s = torch.sin(t).unsqueeze(1)
X = torch.stack([s[i:i + 30] for i in range(1100)])
y = s[30:1130]
m, f = Forecaster(), nn.MSELoss()
opt = torch.optim.Adam(m.parameters(), lr=0.01)
for e in range(80):
    opt.zero_grad(); l = f(m(X), y); l.backward(); opt.step()
print(round(l.item(), 5))
```
