# مسئله بهینه‌سازی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

بهینه‌سازی یعنی یافتن بهترین تصمیم با توجه به محدودیت‌ها. آموزش مدل، زمان‌بندی، مسیریابی و طراحی همگی مسئله بهینه‌سازی‌اند.

$$\min_x\ f(x)\quad s.t.\ \ g_i(x)\le0,\ \ h_j(x)=0$$

| اصطلاح | معنی | مثال |
|---|---|---|
| متغیر تصمیم | آنچه تعیین می‌کنیم | تعداد تولید |
| تابع هدف | آنچه بهتر می‌کنیم | سود |
| قید | محدودیت | بودجه |
| شدنی | جوابی که قیدها را رعایت کند | |
| بهینه محلی | بهتر از همسایه‌ها | |
| بهینه سراسری | بهترین در کل فضا | |
