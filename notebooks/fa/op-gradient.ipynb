# روش‌های مبتنی بر مشتق

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

| روش | به‌روزرسانی | ویژگی |
|---|---|---|
| گرادیان | $x-\eta\nabla f$ | ساده |
| نیوتن | $x-H^{-1}\nabla f$ | هم‌گرایی سریع، هزینه هسین |
| BFGS | تقریب هسین | تعادل خوب |

نیوتن برای تابع درجه دو در یک گام به جواب می‌رسد: $f(x)=x^2-4x$ با $x_0=10$: مشتق ۱۶ و مشتق دوم ۲ است، پس $x_1=10-16/2=2$.



</div>

```python
from scipy.optimize import minimize, rosen

for method in ["Nelder-Mead", "BFGS"]:
    r = minimize(rosen, [-1.2, 1.0], method=method)
    print(method, r.x.round(4), "function calls:", r.nfev)
```
