# نورون مصنوعی و تابع فعال‌سازی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

نورون ورودی‌ها را وزن‌دار جمع می‌کند و از تابع فعال‌سازی می‌گذراند.

$$\hat y=f\Big(\sum_i w_ix_i+b\Big)$$

بدون تابع غیرخطی، چند لایه معادل یک تبدیل خطی می‌شوند. **ReLU** انتخاب رایج لایه‌های پنهان است.

> [توابع فعال‌سازی پرکاربرد.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/activations.png)

*توابع فعال‌سازی پرکاربرد.*

| تابع | بازه | محل استفاده | نکته |
|---|---|---|---|
| ReLU | $[0,\infty)$ | لایه پنهان | سریع؛ خطر مرگ نورون |
| Leaky ReLU | $(-\infty,\infty)$ | لایه پنهان | شیب کوچک در منفی‌ها |
| Sigmoid | $(0,1)$ | خروجی دودویی | محو گرادیان |
| Tanh | $(-1,1)$ | RNN | صفرمرکز |
| Softmax | جمع برابر ۱ | خروجی چندکلاسه | احتمال کلاس‌ها |

## انتخاب تابع خروجی و هزینه
| مسئله | خروجی | هزینه |
|---|---|---|
| رگرسیون | خطی | MSE |
| دودویی | Sigmoid | Binary Cross-Entropy |
| چندکلاسه | Softmax | Cross-Entropy |

</div>

```python
import numpy as np

def sigmoid(z): return 1 / (1 + np.exp(-z))
def relu(z): return np.maximum(z, 0)

x = np.array([0.5, -1.2, 3.0])
w = np.array([0.8, 0.1, -0.4])
z = w @ x + 0.2
print("score:", round(z, 3))
print("sigmoid:", round(sigmoid(z), 3), "relu:", relu(z))
```
