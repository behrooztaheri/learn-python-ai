# بهینه‌ساز و منظم‌سازی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

- **SGD:** گرادیان روی دسته‌های کوچک
- **Momentum:** میانگین گرادیان‌ها برای کاهش نوسان
- **Adam:** گام تطبیقی و نقطه شروع خوب

ابزارهای کنترل بیش‌برازش: **Dropout**، **Weight Decay**، **Early Stopping**، **Batch Normalization** و افزایش داده.

*رفتار SGD، Momentum و Adam در یک دره باریک.*

| بهینه‌ساز | ایده | پیش‌فرض رایج |
|---|---|---|
| SGD | گرادیان دسته کوچک | نرخ ۰٫۱ تا ۰٫۰۰۱ |
| Momentum | حرکت با اینرسی | ضریب ۰٫۹ |
| RMSProp | مقیاس با میانگین مربع گرادیان | نرخ ۰٫۰۰۱ |
| Adam | ترکیب هر دو | نرخ ۰٫۰۰۱ |

$$v_{t+1}=\beta v_t-\eta\nabla J,\quad \theta_{t+1}=\theta_t+v_{t+1}$$

## راهنمای کنترل بیش‌برازش
| ابزار | چگونه کمک می‌کند |
|---|---|
| Dropout | خاموشی تصادفی نورون‌ها |
| Weight Decay | جریمه وزن بزرگ |
| Early Stopping | توقف در بهترین اعتبارسنجی |
| Batch Norm | پایداری آموزش |
| Augmentation | داده متنوع‌تر |

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, stratify=y, random_state=0)
for solver in ["sgd", "adam"]:
    m = make_pipeline(StandardScaler(), MLPClassifier((64,), solver=solver, max_iter=60, random_state=0))
    print(solver, round(m.fit(X_tr, y_tr).score(X_te, y_te), 3))
```
