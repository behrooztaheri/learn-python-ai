# دیکشنری، تاپل و مجموعه

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

- **دیکشنری** هر کلید را به یک مقدار نسبت می‌دهد.
- **تاپل** شبیه لیست است ولی تغییر نمی‌کند.
- **مجموعه** عضو تکراری ندارد و برای عضویت و حذف تکرار سریع است.

## کدام ساختار؟
| ساختار | ترتیب | تغییرپذیر | تکرار | مناسب برای |
|---|---|---|---|---|
| list | دارد | بله | مجاز | دنباله داده |
| tuple | دارد | خیر | مجاز | داده ثابت |
| set | ندارد | بله | مجاز نیست | یکتاسازی و عضویت |
| dict | ترتیب درج | بله | کلید یکتا | نگاشت کلید به مقدار |

```python
text = "to be or not to be"
counts = {}
for w in text.split():
    counts[w] = counts.get(w, 0) + 1
print(counts)
print(sorted(counts.items(), key=lambda kv: kv[1], reverse=True))

a, b = {1, 2, 3}, {2, 3, 4}
print(a & b, a | b, a - b)
```

```python
student = {"name": "Ali", "grade": 18}
student["city"] = "Tehran"
for key, value in student.items():
    print(key, "=", value)
print(student.get("age", "unknown"))

point = (3, 4)
x, y = point
print(x + y)

words = ["a", "b", "a", "c", "b"]
print(set(words), len(set(words)))
```
