# K نزدیک‌ترین همسایه

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

نمونه جدید با رأی k همسایه نزدیک برچسب می‌گیرد. ایده ساده است، ولی مقیاس‌بندی ضروری است. k کوچک به نویز حساس و k بزرگ بیش از حد هموار است.

روی بوم کلیک کن تا نقطه جدید بگذاری و با اسلایدر k مرز را ببین.

</div>

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_wine(return_X_y=True)
for k in [1, 3, 5, 15, 45]:
    raw = cross_val_score(KNeighborsClassifier(k), X, y, cv=5).mean()
    scaled = cross_val_score(make_pipeline(StandardScaler(), KNeighborsClassifier(k)), X, y, cv=5).mean()
    print(f"k={k:2d} raw={raw:.3f} scaled={scaled:.3f}")
```
