# عدد و رشته

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

عملگرهای عددی: `+` و `-` و `*` و `/` (تقسیم اعشاری)، `//` (تقسیم صحیح)، `%` (باقی‌مانده) و `**` (توان).

رشته دنباله‌ای از کاراکترهاست و شماره‌گذاری آن از صفر شروع می‌شود. **f-string** ساده‌ترین راه قرار دادن مقدار داخل متن است.

با آزمایشگاه زیر ببین برش `[start:stop:step]` دقیقاً چه چیزی را انتخاب می‌کند.

## متدهای پرکاربرد رشته
| متد | کار | مثال | نتیجه |
|---|---|---|---|
| `strip` | حذف فاصله دو سر | `" a ".strip()` | `"a"` |
| `lower` | حروف کوچک | `"Ab".lower()` | `"ab"` |
| `split` | شکستن به لیست | `"a,b".split(",")` | `["a","b"]` |
| `join` | چسباندن لیست | `"-".join(["a","b"])` | `"a-b"` |
| `replace` | جایگزینی | `"cat".replace("c","b")` | `"bat"` |
| `find` | محل زیررشته | `"data".find("t")` | `2` |

## اولویت عملگرها
به‌ترتیب: پرانتز، توان، ضرب و تقسیم، جمع و تفریق. وقتی تردید داری پرانتز بگذار.

> اعداد اعشاری در دودویی دقیق ذخیره نمی‌شوند. برای مقایسه از `math.isclose` استفاده کن.



</div>

```python
import math

print(0.1 + 0.2 == 0.3, 0.1 + 0.2)
print(math.isclose(0.1 + 0.2, 0.3))
print(2 + 3 * 4, (2 + 3) * 4, 2 ** 3 ** 2)
print("-".join(["a", "b", "c"]), "  hi ".strip())
```

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)

text = "Machine Learning"
print(text[0], text[-1], text[0:7])
print(text.upper(), text.replace("Machine", "Deep"))
print(len(text), text.split())

score = 18.456
print(f"Score: {score:.1f} out of 20")
```
