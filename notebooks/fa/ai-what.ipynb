# هوش مصنوعی چیست؟

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

**هوش مصنوعی** حوزه ساخت سیستم‌هایی است که کارهای نیازمند هوش را انجام می‌دهند. **یادگیری ماشین** زیرمجموعه آن است و به‌جای قانون دستی، الگو را از داده یاد می‌گیرد. **یادگیری عمیق** زیرمجموعه یادگیری ماشین است و از شبکه‌های عصبی چندلایه استفاده می‌کند.

## کاربرد در رشته‌های مختلف
- پزشکی: کمک به تشخیص از روی تصویر
- کشاورزی: تشخیص بیماری گیاه
- مالی: شناسایی تقلب
- زبان: ترجمه و دستیارهای گفت‌وگو

## چه زمانی مناسب نیست؟
اگر قانون ساده کافی است، داده کافی نداری، یا خطا پیامد سنگین دارد و توضیح‌پذیری لازم است، باید با احتیاط تصمیم بگیری.
