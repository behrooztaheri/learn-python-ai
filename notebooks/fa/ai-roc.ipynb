## منحنی ROC
روی هر آستانه نرخ مثبت واقعی را برحسب نرخ مثبت کاذب رسم می‌کند.

$$TPR=\frac{TP}{TP+FN}\qquad FPR=\frac{FP}{FP+TN}$$

**AUC** مساحت زیر منحنی است و احتمال آن است که یک نمونه مثبت تصادفی امتیاز بیشتری از یک نمونه منفی تصادفی بگیرد. مقدار ۰٫۵ معادل حدس تصادفی است.

> [منحنی ROC و منحنی Precision-Recall روی داده واقعی سرطان سینه.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/roc_pr.png)

*منحنی ROC و منحنی Precision-Recall روی داده واقعی سرطان سینه.*

## ROC یا PR؟
| موقعیت | پیشنهاد |
|---|---|
| کلاس‌ها تقریباً متوازن | ROC |
| کلاس مثبت بسیار کم | PR |



</div>

```python
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, precision_score, recall_score, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_breast_cancer(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, stratify=y, random_state=0)
m = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(X_tr, y_tr)
p = m.predict_proba(X_te)[:, 1]
print("ROC AUC:", round(roc_auc_score(y_te, p), 4), "AP:", round(average_precision_score(y_te, p), 4))
for t in [0.2, 0.5, 0.8, 0.95]:
    pred = p >= t
    print(t, "precision", round(precision_score(y_te, pred), 3), "recall", round(recall_score(y_te, pred), 3))
```
