# Transformer و مدل‌های زبانی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

در **Self-Attention** هر توکن به همه توکن‌ها نگاه می‌کند.

$$Attention(Q,K,V)=softmax\Big(\frac{QK^T}{\sqrt{d_k}}\Big)V$$

مدل‌های زبانی بزرگ روی این معماری ساخته شده‌اند: متن به توکن تبدیل می‌شود و مدل توکن بعدی را احتمالاتی پیش‌بینی می‌کند.

> مدل زبانی ممکن است پاسخ نادرست ولی مطمئن تولید کند. نتیجه مهم را همیشه بررسی کن.

## اجزای Transformer
| جزء | نقش |
|---|---|
| Embedding | تبدیل توکن به بردار |
| Positional Encoding | افزودن اطلاعات ترتیب |
| Multi-Head Attention | چند نوع توجه موازی |
| Feed Forward | پردازش هر توکن |
| Residual و Layer Norm | آموزش پایدار شبکه عمیق |

در توجه، **Query** می‌پرسد، **Key** معرفی می‌کند و **Value** محتوا را می‌دهد. هرچه Query و Key شبیه‌تر باشند، وزن بیشتری می‌گیرند.

</div>

```python
import numpy as np

# Scaled dot-product attention on three tokens
Q = K = V = np.array([[1.0, 0.0], [0.9, 0.1], [0.0, 1.0]])
scores = Q @ K.T / np.sqrt(K.shape[1])
exp = np.exp(scores - scores.max(axis=1, keepdims=True))
weights = exp / exp.sum(axis=1, keepdims=True)
print(weights.round(3))
print((weights @ V).round(3))
```
