## راه‌حل‌ها
انتخاب ویژگی، استخراج ویژگی مثل PCA، منظم‌سازی، داده بیشتر و دانش تخصصی.

## چقدر نمونه لازم است؟
با ۱۰ نقطه در هر محور:

| تعداد بعد | تعداد نمونه |
|---|---|
| ۱ | 10 |
| ۲ | 100 |
| ۳ | 1,000 |
| ۱۰ | 10,000,000,000 |

برای پوشاندن ۱۰٪ حجم یک مکعب واحد در d بعد، ضلع $0.1^{1/d}$ لازم است: در ۱۰ بعد ۰٫۷۹ و در ۱۰۰ بعد ۰٫۹۸. یعنی «همسایگی» تقریباً کل بازه را می‌گیرد.

```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_wine(return_X_y=True)
rng = np.random.default_rng(0)
for extra in [0, 20, 100, 500]:
    Z = np.hstack([StandardScaler().fit_transform(X), rng.normal(size=(len(X), extra))])
    acc = cross_val_score(KNeighborsClassifier(5), Z, y, cv=5).mean()
    print(f"{extra:4d} useless features -> accuracy {acc:.3f}")
```
