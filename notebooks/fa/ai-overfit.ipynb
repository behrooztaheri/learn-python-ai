# کم‌برازش و بیش‌برازش

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

- **کم‌برازش:** مدل بیش از حد ساده است و خطای آموزش و آزمون هر دو زیادند.
- **بیش‌برازش:** مدل نویز را حفظ کرده؛ خطای آموزش کم و خطای آزمون زیاد است.

با حرکت اسلایدر درجه را تغییر بده و ببین از جایی به بعد خطای آزمون بالا می‌رود.

## موازنه بایاس و واریانس
خطای مورد انتظار روی داده جدید از سه بخش ساخته می‌شود:

$$E[(y-\hat f)^2]=Bias^2+Variance+\sigma^2$$

- **بایاس:** خطای ناشی از ساده بودن مدل.
- **واریانس:** حساسیت مدل به نمونه آموزش.
- $\sigma^2$: نویزی که هیچ مدلی آن را حذف نمی‌کند.
