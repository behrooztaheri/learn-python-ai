# Git و GitHub در یک نگاه

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

Git نسخه‌های کد را ذخیره می‌کند و GitHub میزبانی آنلاین و همکاری را ممکن می‌کند.

- `git add` تغییرها را آماده می‌کند.
- `git commit` یک نقطه ذخیره می‌سازد.
- `git push` آن را روی GitHub می‌فرستد.
- `git pull` تغییرها را از GitHub می‌گیرد.

> رمزها و فایل‌های حجیم را در مخزن نگذار و آن‌ها را در `.gitignore` بنویس.

</div>

```bash
git init
git add .
git commit -m "First version"
git branch -M main
git remote add origin https://github.com/USERNAME/REPOSITORY.git
git push -u origin main
```
