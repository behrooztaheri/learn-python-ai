# معیارهای رگرسیون

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

$$MAE=\frac{1}{n}\sum|y_i-\hat y_i|\qquad RMSE=\sqrt{\frac{1}{n}\sum(y_i-\hat y_i)^2}$$

$$R^2=1-\frac{\sum(y_i-\hat y_i)^2}{\sum(y_i-\bar y)^2}$$

- **MAE** میانگین اندازه خطا با واحد خروجی است.
- **RMSE** به خطای بزرگ حساس‌تر است.
- **R²** سهم تغییرات توضیح‌داده‌شده است و می‌تواند منفی شود.

> R² برابر ۰٫۹ یعنی ۹۰٪ تغییرات توضیح داده شده؛ نه اینکه ۹۰٪ پیش‌بینی‌ها درست باشند.

## مثال کامل با چهار خانه
| خانه | واقعی y | پیش‌بینی ŷ | خطا | قدرمطلق | مربع |
|---|---|---|---|---|---|
| ۱ | 100 | 110 | −10 | 10 | 100 |
| ۲ | 150 | 140 | 10 | 10 | 100 |
| ۳ | 200 | 210 | −10 | 10 | 100 |
| ۴ | 250 | 240 | 10 | 10 | 100 |

- میانگین قدرمطلق خطا: $MAE=40/4=10$
- میانگین مربعات: $MSE=400/4=100$
- ریشه آن: $RMSE=\sqrt{100}=10$
- میانگین واقعی ۱۷۵ است، پس $\sum(y-\bar y)^2=12500$ و $R^2=1-400/12500=0.968$

## کدام معیار؟
| معیار | بهتر | واحد | چه زمانی |
|---|---|---|---|
| MAE | نزدیک صفر | واحد خروجی | بیان ساده خطا |
| MSE | نزدیک صفر | مربع واحد | تابع هزینه آموزش |
| RMSE | نزدیک صفر | واحد خروجی | خطای بزرگ پرهزینه |
| R² | نزدیک ۱ | بدون واحد | مقایسه با پیش‌بینی میانگین |
| MAPE | نزدیک صفر | درصد | خروجی‌های دور از صفر |

> اگر R² منفی شد، مدل از پیش‌بینی ساده میانگین هم بدتر است.

</div>

```python
import numpy as np

y = np.array([100, 150, 200, 250])
p = np.array([110, 140, 210, 240])
e = y - p

print("MAE :", np.mean(np.abs(e)))
print("RMSE:", np.sqrt(np.mean(e ** 2)))
print("R2  :", 1 - np.sum(e ** 2) / np.sum((y - y.mean()) ** 2))
```
