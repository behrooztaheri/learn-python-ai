# PCA و کاهش بعد

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

PCA جهت‌هایی با بیشترین واریانس پیدا می‌کند و داده را روی چند جهت اول تصویر می‌کند. برای نمایش، فشرده‌سازی و کاهش هم‌خطی مفید است. PCA برچسب‌ها را نمی‌بیند.

زاویه محور را تغییر بده و ببین در کدام جهت واریانس بیشینه می‌شود.

> [داده ۶۴بعدی ارقام و درصد واریانس تجمعی.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/pca.png)

*داده ۶۴بعدی ارقام و درصد واریانس تجمعی.*

## مراحل
- داده را میانگین صفر و معمولاً استاندارد کن.
- ماتریس کوواریانس را بساز.
- بردارها و مقدارهای ویژه را پیدا کن.
- مقدار ویژه‌ها را مرتب و k بردار اول را انتخاب کن.
- داده را روی آن‌ها تصویر کن.

## مثال دو بعدی
اگر کوواریانس $\begin{bmatrix}2&1\\1&2\end{bmatrix}$ باشد، مقدار ویژه‌ها ۳ و ۱ هستند. پس مؤلفه اول $3/(3+1)=75\%$ واریانس را دارد.



</div>

```python
import numpy as np

values, vectors = np.linalg.eigh(np.array([[2.0, 1.0], [1.0, 2.0]]))
order = values.argsort()[::-1]
print("eigenvalues:", values[order])
print("first direction:", vectors[:, order[0]].round(3))
print("variance ratio:", (values[order] / values.sum()).round(3))
```

```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X = StandardScaler().fit_transform(load_digits().data)
p = PCA().fit(X)
cum = np.cumsum(p.explained_variance_ratio_)
print("Components for 95%:", int(np.searchsorted(cum, 0.95) + 1), "of", X.shape[1])
print("First 3 ratios:", p.explained_variance_ratio_[:3].round(3))
```
