# بازی مجموع صفر و Minimax

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

در بازی **مجموع صفر** سود یکی زیان دیگری است. فون نویمان نشان داد ارزش بازی با راهبرد مختلط یکتا است:

$$\max_p\min_q\ p^TAq=\min_q\max_p\ p^TAq$$

## مثال
برای ماتریس $A=\begin{bmatrix}3&-1\\-2&4\end{bmatrix}$ با برابر قرار دادن پاداش دو ستون: $3p-2(1-p)=-p+4(1-p)$، پس $p=0.6$ و ارزش بازی ۱ است.

این مسئله با برنامه‌ریزی خطی هم حل می‌شود:



</div>

```python
import numpy as np
from scipy.optimize import linprog

def solve_zero_sum(A):
    A = np.array(A, float)
    m, n = A.shape
    c = [0] * m + [-1]                  # maximize the value v
    A_ub = np.c_[-A.T, np.ones(n)]
    A_eq = [[1] * m + [0]]
    r = linprog(c, A_ub=A_ub, b_ub=np.zeros(n), A_eq=A_eq, b_eq=[1], bounds=[(0, 1)] * m + [(None, None)])
    return r.x[:m].round(3), round(r.x[-1], 3)

print(solve_zero_sum([[3, -1], [-2, 4]]))
print("Rock-paper-scissors:", solve_zero_sum([[0, -1, 1], [1, 0, -1], [-1, 1, 0]]))
```

## جستجوی Minimax در بازی‌ها
در بازی‌های نوبتی مثل دوز، درخت حالت‌ها پیمایش می‌شود. **هرس آلفا-بتا** شاخه‌های بی‌فایده را حذف می‌کند.



</div>

```python
from functools import lru_cache

LINES = [(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]

def winner(b):
    for i, j, k in LINES:
        if b[i] != "." and b[i] == b[j] == b[k]:
            return b[i]
    return None

@lru_cache(maxsize=None)
def value(b, player):
    w = winner(b)
    if w:
        return 1 if w == "X" else -1
    if "." not in b:
        return 0
    other = "O" if player == "X" else "X"
    scores = [value(b[:i] + player + b[i+1:], other) for i, c in enumerate(b) if c == "."]
    return max(scores) if player == "X" else min(scores)

print("Value of tic-tac-toe with perfect play:", value("." * 9, "X"))
```
