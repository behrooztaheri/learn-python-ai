# پروژه کامل: از داده تا گزارش

مراحل یک پروژه واقعی: تعریف مسئله، شناخت داده، خط لوله، انتخاب مدل با اعتبارسنجی، ارزیابی نهایی و گزارش صادقانه.

> داده آزمون را فقط در پایان استفاده کن. اگر بعد از دیدن نتیجه آزمون مدل را تغییر بدهی، برآوردت خوش‌بینانه می‌شود.

**تمرین:** `load_wine` را جایگزین کن و یک مدل جدید به فهرست اضافه کن.

</div>

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X, y = load_breast_cancer(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)

models = {
    "logistic": make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
    "svm": make_pipeline(StandardScaler(), SVC()),
    "forest": RandomForestClassifier(200, random_state=42),
}
scores = {n: cross_val_score(m, X_tr, y_tr, cv=5, scoring="f1").mean() for n, m in models.items()}
print({n: round(s, 4) for n, s in scores.items()})

best = max(scores, key=scores.get)
final = models[best].fit(X_tr, y_tr)
print("Best model:", best)
print(classification_report(y_te, final.predict(X_te), digits=3))
ConfusionMatrixDisplay.from_estimator(final, X_te, y_te, cmap="Purples")
plt.show()
```
