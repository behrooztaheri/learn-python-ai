رگرسیون خطی خروجی را با ترکیب خطی ورودی‌ها تخمین می‌زند و وزن‌ها را طوری می‌یابد که مجموع مربع خطا کمینه شود.

$$\hat y=b_0+b_1x_1+\dots+b_px_p$$

روی بوم کلیک کن تا نقطه اضافه شود و ببین خط چگونه تغییر می‌کند. یک نقطه پرت هم بگذار!

</div>

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
X, y = load_diabetes(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0)

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

model = LinearRegression().fit(X_tr, y_tr)
p = model.predict(X_te)
print("RMSE:", round(np.sqrt(mean_squared_error(y_te, p)), 2))
print("R2  :", round(r2_score(y_te, p), 3))
```
