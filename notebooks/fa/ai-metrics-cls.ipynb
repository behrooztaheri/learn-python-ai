# معیارهای طبقه‌بندی

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

ماتریس اغتشاش چهار حالت دارد: TP و TN درست، FP هشدار کاذب و FN مورد جاافتاده.

$$Precision=\frac{TP}{TP+FP}\qquad Recall=\frac{TP}{TP+FN}$$

$$F_1=\frac{2\cdot Precision\cdot Recall}{Precision+Recall}$$

- وقتی هشدار کاذب گران است، Precision مهم است.
- وقتی جاافتادن گران است، Recall مهم است.
- در داده نامتوازن Accuracy گمراه‌کننده است.

اعداد را تغییر بده و اثرش را روی معیارها ببین.
