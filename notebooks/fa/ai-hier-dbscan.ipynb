# خوشه‌بندی سلسله‌مراتبی و DBSCAN

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

- **سلسله‌مراتبی:** خوشه‌ها را مرحله‌به‌مرحله ادغام می‌کند و نتیجه را با دندروگرام نشان می‌دهد.
- **DBSCAN:** ناحیه‌های چگال را خوشه می‌داند، شکل دلخواه را پیدا می‌کند و نویز را جدا می‌کند. `eps` و `min_samples` پارامترهای اصلی‌اند.

</div>

```python
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering, DBSCAN, KMeans
from sklearn.datasets import make_moons

X, _ = make_moons(300, noise=0.06, random_state=0)
models = [("KMeans", KMeans(2, n_init=10, random_state=0)), ("Ward", AgglomerativeClustering(2)), ("DBSCAN", DBSCAN(eps=0.2, min_samples=5))]
fig, ax = plt.subplots(1, 3, figsize=(10, 3))
for a, (name, m) in zip(ax, models):
    a.scatter(X[:, 0], X[:, 1], c=m.fit_predict(X), s=12, cmap="viridis")
    a.set_title(name)
plt.show()
```
