ضرب داخلی پایه مدل‌های خطی است:

$$w^Tx=\sum_{i=1}^{n}w_ix_i$$

```python
import numpy as np

w = np.array([0.5, -1.0, 2.0])
x = np.array([4.0, 1.0, 3.0])
print("dot product:", w @ x)

f = lambda t: t ** 2 + 3 * t
t, h = 2.0, 1e-5
print("numerical derivative at 2:", round((f(t + h) - f(t - h)) / (2 * h), 4))

data = np.array([2, 4, 4, 4, 5, 5, 7, 9])
print("mean:", data.mean(), "std:", data.std())
```
