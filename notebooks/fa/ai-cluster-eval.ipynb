# ارزیابی خوشه‌بندی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

در خوشه‌بندی برچسب واقعی نداریم، پس کیفیت را با معیارهای داخلی می‌سنجیم.

| معیار | فرمول | بهتر |
|---|---|---|
| WCSS | $\sum_j\sum_{x\in C_j}\Vert x-\mu_j\Vert ^2$ | کمتر |
| Silhouette | $s(i)=\frac{b(i)-a(i)}{\max(a(i),b(i))}$ | نزدیک ۱ |
| Davies-Bouldin | $\frac{1}{k}\sum_i\max_{j\ne i}\frac{S_i+S_j}{M_{ij}}$ | کمتر |

## Silhouette با یک عدد
$a$ میانگین فاصله نقطه تا هم‌خوشه‌ای‌هاست و $b$ میانگین فاصله تا نزدیک‌ترین خوشه دیگر. اگر $a=2$ و $b=5$ باشد:

$$s=\frac{5-2}{5}=0.6$$

| مقدار s | معنی |
|---|---|
| نزدیک ۱ | نقطه در خوشه درست است |
| نزدیک ۰ | نزدیک مرز دو خوشه |
| منفی | احتمالاً در خوشه اشتباه |

> [روش آرنج، میانگین Silhouette و نمودار Silhouette هر نمونه روی داده Iris.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/cluster_eval.png)

*روش آرنج، میانگین Silhouette و نمودار Silhouette هر نمونه روی داده Iris.*



</div>

```python
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.metrics import adjusted_rand_score, davies_bouldin_score, silhouette_score
from sklearn.preprocessing import StandardScaler

d = load_iris()
X = StandardScaler().fit_transform(d.data)
for k in [2, 3, 4]:
    l = KMeans(k, n_init=10, random_state=0).fit_predict(X)
    print(k, round(silhouette_score(X, l), 3), round(davies_bouldin_score(X, l), 3), round(adjusted_rand_score(d.target, l), 3))
```
