# مدیریت خطا و خواندن پیام خطا

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

پیام خطا دشمن نیست. **آخرین خط** آن نوع خطا را می‌گوید و خط‌های بالاتر محل آن را نشان می‌دهند.

خطاهای رایج: `SyntaxError` و `NameError` و `TypeError` و `IndexError` و `KeyError`.

با `try` و `except` خطای قابل پیش‌بینی را مدیریت کن، ولی همه خطاها را کورکورانه پنهان نکن.

## خطاهای رایج و درمان
| خطا | معنی | معمولاً |
|---|---|---|
| `SyntaxError` | جمله‌بندی نادرست | دونقطه یا پرانتز جا افتاده |
| `IndentationError` | تورفتگی نادرست | ترکیب نادرست فاصله‌ها |
| `NameError` | نام تعریف نشده | غلط تایپی |
| `TypeError` | نوع نامناسب | جمع متن و عدد |
| `IndexError` | شماره خارج از بازه | لیست کوتاه‌تر است |
| `KeyError` | کلید پیدا نشد | از `get` استفاده کن |
| `FileNotFoundError` | فایل نیست | مسیر نادرست |

## روش عیب‌یابی
- آخرین خط پیام را بخوان.
- شماره خطی را که از کد خودت است پیدا کن.
- مقدار متغیرها را با `print` یا Debugger ببین.
- مسئله را به کوچک‌ترین نمونه تکرارپذیر برسان.

</div>

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))

for text in ["12", "abc"]:
    try:
        print(int(text) * 2)
    except ValueError as err:
        print("Bad value:", err)
```
