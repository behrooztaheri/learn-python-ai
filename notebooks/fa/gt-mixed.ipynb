# راهبرد مختلط

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

گاهی بازیکن باید تصادفی بازی کند تا قابل پیش‌بینی نباشد. در تعادل مختلط، احتمال هر بازیکن طوری است که **طرف مقابل بین گزینه‌هایش بی‌تفاوت شود**.

## مثال: نبرد زوج‌ها
| | ستون A | ستون B |
|---|---|---|
| **سطر A** | (2, 1) | (0, 0) |
| **سطر B** | (0, 0) | (1, 2) |

بازیکن ۱ با احتمال p گزینه A را بازی می‌کند. بازیکن ۲ باید بی‌تفاوت شود:

$$1\cdot p=2(1-p)\Rightarrow p=\tfrac{2}{3}$$

به همین ترتیب بازیکن ۲ با $q$ بازیکن ۱ را بی‌تفاوت می‌کند: $2q=1-q$، پس $q=\tfrac{1}{3}$.



</div>

```python
from fractions import Fraction as F

a, b, c, d = F(1), F(0), F(0), F(2)   # player 2's payoffs for rows A,B and columns A,B
p = (d - c) / (a - b - c + d)
A, C = F(2), F(1)
q = C / (A + C)
print("p =", p, "q =", q)
print("Expected payoff of player 1:", A * q)
```
