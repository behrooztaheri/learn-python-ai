# تعادل نش

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

در **تعادل نش** هیچ بازیکنی با تغییر یک‌طرفه راهبردش سود بیشتری نمی‌برد. جان نش ثابت کرد هر بازی متناهی دست‌کم یک تعادل (شاید مختلط) دارد.

## روش یافتن تعادل خالص
برای هر ستون، بهترین پاسخ بازیکن ۱ و برای هر سطر، بهترین پاسخ بازیکن ۲ را علامت بزن. خانه‌ای که هر دو علامت را دارد تعادل نش است.

| بازی | تعادل خالص | ویژگی |
|---|---|---|
| معمای زندانی | (اعتراف، اعتراف) | یک تعادل ناکارا |
| نبرد زوج‌ها | دو تعادل | مسئله هماهنگی |
| شکار گوزن | دو تعادل | امن در برابر پربازده |
| سکه‌بازی | ندارد | فقط مختلط |



</div>

```python
import numpy as np

def pure_nash(A, B):
    """Cells where both players are playing a best response."""
    out = []
    for i in range(A.shape[0]):
        for j in range(A.shape[1]):
            if A[i, j] == A[:, j].max() and B[i, j] == B[i, :].max():
                out.append((i, j))
    return out

prisoners_A = np.array([[-1, -3], [0, -2]])
print("Prisoner's dilemma:", pure_nash(prisoners_A, prisoners_A.T))
sexes_A = np.array([[2, 0], [0, 1]])
sexes_B = np.array([[1, 0], [0, 2]])
print("Battle of the sexes:", pure_nash(sexes_A, sexes_B))
```
