# برنامه‌ریزی خطی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

در برنامه‌ریزی خطی هدف و قیدها خطی‌اند. جواب بهینه روی یک گوشه ناحیه شدنی است.

## مثال
کارخانه‌ای دو محصول دارد و می‌خواهد سود را بیشینه کند:

$$\max\ 3x+5y\quad s.t.\ x\le4,\ 2y\le12,\ 3x+2y\le18,\ x,y\ge0$$

| گوشه | سود $3x+5y$ |
|---|---|
| (0,0) | 0 |
| (4,0) | 12 |
| (4,3) | 27 |
| (2,6) | **36** |
| (0,6) | 30 |



</div>

```python
from scipy.optimize import linprog

result = linprog(c=[-3, -5], A_ub=[[1, 0], [0, 2], [3, 2]], b_ub=[4, 12, 18], bounds=[(0, None), (0, None)])
print("x, y =", result.x.round(3), "profit =", round(-result.fun, 3))
```
