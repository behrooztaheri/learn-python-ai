# ازدحام ذرات (PSO)

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

هر ذره موقعیت و سرعت دارد و به بهترین تجربه خودش و بهترین تجربه گروه کشیده می‌شود.

$$v\leftarrow wv+c_1r_1(p_{best}-x)+c_2r_2(g_{best}-x),\quad x\leftarrow x+v$$

| پارامتر | نقش | مقدار رایج |
|---|---|---|
| $w$ | اینرسی | 0.4 تا 0.9 |
| $c_1$ | تجربه شخصی | 1.5 تا 2 |
| $c_2$ | تجربه جمعی | 1.5 تا 2 |

</div>

```python
import numpy as np

def rastrigin(x):
    x = np.atleast_2d(x)
    return 10 * x.shape[1] + np.sum(x ** 2 - 10 * np.cos(2 * np.pi * x), axis=1)

LOW, HIGH, DIM = -5.12, 5.12, 2
rng = np.random.default_rng(1)

x = rng.uniform(LOW, HIGH, (40, DIM))
v = np.zeros_like(x)
pbest, pval = x.copy(), rastrigin(x)
for it in range(150):
    g = pbest[pval.argmin()]
    r1, r2 = rng.random((2, 40, DIM))
    v = 0.7 * v + 1.5 * r1 * (pbest - x) + 1.5 * r2 * (g - x)
    x = np.clip(x + v, LOW, HIGH)
    f = rastrigin(x)
    better = f < pval
    pbest[better], pval[better] = x[better], f[better]
print("best:", pbest[pval.argmin()].round(4), "value:", round(float(pval.min()), 5))
```
