# ماشین بردار پشتیبان

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

SVM مرزی با بیشترین حاشیه پیدا می‌کند و فقط **بردارهای پشتیبان** آن را تعیین می‌کنند. ترفند کرنل مرز غیرخطی می‌سازد. `C` تحمل خطا و `gamma` دامنه اثر هر نمونه را تنظیم می‌کند.

</div>

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, stratify=y, random_state=0)
grid = GridSearchCV(make_pipeline(StandardScaler(), SVC()), {"svc__C": [1, 10], "svc__gamma": ["scale", 0.01]}, cv=3).fit(X_tr, y_tr)
print(grid.best_params_)
print("Test accuracy:", round(grid.score(X_te, y_te), 3))
```
