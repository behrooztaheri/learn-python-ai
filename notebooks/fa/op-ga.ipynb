# الگوریتم ژنتیک

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

الگوریتم ژنتیک از تکامل الهام گرفته است.

| مرحله | کار |
|---|---|
| جمعیت اولیه | چند جواب تصادفی |
| ارزیابی | شایستگی هر جواب |
| انتخاب | جواب‌های بهتر شانس بیشتری دارند |
| تقاطع | ترکیب دو والد |
| جهش | تغییر کوچک تصادفی |
| نخبه‌گرایی | حفظ بهترین جواب |



</div>

```python
import numpy as np

def rastrigin(x):
    x = np.atleast_2d(x)
    return 10 * x.shape[1] + np.sum(x ** 2 - 10 * np.cos(2 * np.pi * x), axis=1)

LOW, HIGH, DIM = -5.12, 5.12, 2
rng = np.random.default_rng(1)

pop = rng.uniform(LOW, HIGH, (60, DIM))
for gen in range(120):
    fit = rastrigin(pop)
    elite = pop[fit.argmin()].copy()
    a, b = rng.integers(0, 60, (2, 60))            # tournament selection
    parents = pop[np.where(fit[a] < fit[b], a, b)]
    alpha = rng.random((60, 1))                    # blend crossover
    children = alpha * parents + (1 - alpha) * parents[::-1]
    mask = rng.random(children.shape) < 0.15       # mutation
    children[mask] += rng.normal(0, 0.4, mask.sum())
    pop = np.clip(children, LOW, HIGH)
    pop[0] = elite                                 # elitism
best = pop[rastrigin(pop).argmin()]
print("best:", best.round(4), "value:", rastrigin(best).round(5))
```

## آزمایشگاه الگوریتم ژنتیک
هر نقطه نارنجی یک کروموزوم ۸ بیتی است (عددی بین ۰ تا ۲۵۵). «نسل بعد» را بزن و ببین جمعیت چطور به سمت قله می‌رود. احتمال جهش را صفر کن: جمعیت زود یکسان می‌شود و ممکن است در قله محلی گیر کند.
