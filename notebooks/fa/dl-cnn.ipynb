کرنل کوچکی روی تصویر می‌لغزد و **نقشه ویژگی** می‌سازد. **اشتراک وزن** تعداد پارامترها را کم می‌کند. Pooling اندازه را کم می‌کند و لایه‌های عمیق‌تر الگوهای بزرگ‌تر را می‌بینند.

```python
import torch, torch.nn as nn

model = nn.Sequential(
    nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
    nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
    nn.Flatten(), nn.Linear(32 * 7 * 7, 10),
)
x = torch.zeros(4, 1, 28, 28)
print(model(x).shape, sum(p.numel() for p in model.parameters()))
```
