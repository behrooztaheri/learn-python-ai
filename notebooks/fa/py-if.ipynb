# شرط‌ها

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

با `if`، `elif` و `else` برنامه تصمیم می‌گیرد. **تورفتگی** (چهار فاصله) در پایتون بخشی از دستور زبان است و بدنه شرط را مشخص می‌کند.

عملگرهای مقایسه: `==` و `!=` و `<` و `>` و `<=` و `>=`. برای ترکیب شرط‌ها: `and` و `or` و `not`.

</div>

```python
grade = 15

if grade >= 17:
    result = "Excellent"
elif grade >= 10:
    result = "Passed"
else:
    result = "Failed"

print(result)
print("In range:", 0 <= grade <= 20)
```
