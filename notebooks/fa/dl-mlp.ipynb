# شبکه چندلایه و پس‌انتشار

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

چند لایه نورون یک **MLP** می‌سازند. آموزش چهار مرحله دارد:

- عبور رو به جلو و محاسبه خروجی
- محاسبه خطا
- پس‌انتشار و محاسبه گرادیان با قاعده زنجیره‌ای
- به‌روزرسانی وزن‌ها

$$\frac{\partial J}{\partial w}=\frac{\partial J}{\partial \hat y}\cdot\frac{\partial \hat y}{\partial h}\cdot\frac{\partial h}{\partial w}$$

در آزمایشگاه یک شبکه واقعی را آموزش بده و مرز تصمیمش را ببین.

```python
import numpy as np

# XOR: a problem one neuron cannot solve
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], float)
y = np.array([[0], [1], [1], [0]], float)

rng = np.random.default_rng(1)
W1, b1 = rng.normal(size=(2, 4)), np.zeros((1, 4))
W2, b2 = rng.normal(size=(4, 1)), np.zeros((1, 1))
sig = lambda z: 1 / (1 + np.exp(-z))

for epoch in range(5001):
    h = np.tanh(X @ W1 + b1)                # forward
    p = sig(h @ W2 + b2)
    loss = -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))
    g2 = p - y                              # backward
    dW2, db2 = h.T @ g2 / 4, g2.mean(0, keepdims=True)
    g1 = (g2 @ W2.T) * (1 - h ** 2)
    dW1, db1 = X.T @ g1 / 4, g1.mean(0, keepdims=True)
    W1 -= 0.5 * dW1; b1 -= 0.5 * db1        # update
    W2 -= 0.5 * dW2; b2 -= 0.5 * db2
    if epoch % 1000 == 0:
        print(epoch, round(float(loss), 4))

print(p.round(3).ravel())
```
