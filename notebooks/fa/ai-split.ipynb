# تقسیم داده و اعتبارسنجی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

مدل باید روی داده‌ای ارزیابی شود که ندیده است.

- **آموزش:** یادگیری پارامترها
- **اعتبارسنجی:** انتخاب مدل و تنظیم ابرپارامتر
- **آزمون:** ارزیابی نهایی، فقط یک بار

در **k-Fold** داده به k بخش تقسیم می‌شود و هر بخش یک بار نقش اعتبارسنجی را دارد. در سری زمانی باید ترتیب زمان حفظ شود.

## کدام روش؟
| روش | چه زمانی |
|---|---|
| `train_test_split` | داده زیاد و بررسی سریع |
| `KFold` | داده متوسط |
| `StratifiedKFold` | طبقه‌بندی، به‌ویژه نامتوازن |
| `GroupKFold` | چند نمونه از یک بیمار یا کاربر |
| `TimeSeriesSplit` | داده زمانی |

## تنظیم ابرپارامتر
`GridSearchCV` همه ترکیب‌ها و `RandomizedSearchCV` نمونه‌ای تصادفی را می‌آزماید. جستجوی تصادفی وقتی پارامتر زیاد است معمولاً به‌صرفه‌تر است.

> تقسیم تصادفی برای سری زمانی نادرست است، چون مدل از آینده اطلاعات می‌گیرد.

</div>

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_iris(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)

model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=500))
scores = cross_val_score(model, X_tr, y_tr, cv=5)
print("CV accuracy:", scores.round(3), "mean:", scores.mean().round(3))
print("Test accuracy:", round(model.fit(X_tr, y_tr).score(X_te, y_te), 3))
```
