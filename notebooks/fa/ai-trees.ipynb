درخت با پرسش‌های پی‌درپی داده را تقسیم می‌کند و معیارهایی مثل **Gini** خلوص گروه‌ها را می‌سنجند.

$$Gini=1-\sum_k p_k^2$$

درخت تفسیرپذیر است و به مقیاس‌بندی نیاز ندارد، ولی بدون محدودیت عمق به‌راحتی بیش‌برازش می‌کند.

</div>

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, export_text

d = load_iris()
m = DecisionTreeClassifier(max_depth=2, random_state=0).fit(d.data, d.target)
print(export_text(m, feature_names=list(d.feature_names)))
```
