# Pandas

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

Pandas برای داده جدولی است. `DataFrame` یک جدول و `Series` یک ستون است.

- خواندن: `read_csv` و `read_excel`
- شناخت داده: `head` و `info` و `describe`
- انتخاب: `loc` با نام و `iloc` با شماره
- مقدار گمشده: `isna` و `fillna` و `dropna`
- گروه‌بندی: `groupby`

## عملیات کلیدی
| کار | کد |
|---|---|
| انتخاب ستون | `df["a"]` |
| فیلتر | `df[df.a > 3]` |
| ستون جدید | `df["c"] = df.a * 2` |
| مرتب‌سازی | `df.sort_values("a")` |
| گروه‌بندی | `df.groupby("g").a.mean()` |
| ادغام | `pd.merge(a, b, on="id")` |
| تابع روی ستون | `df.a.apply(f)` |



</div>

```python
import pandas as pd

students = pd.DataFrame({"id": [1, 2, 3], "name": ["Ali", "Sara", "Reza"]})
grades = pd.DataFrame({"id": [1, 1, 2, 3], "course": ["AI", "Math", "AI", "AI"], "mark": [18, 15, 20, 12]})

df = pd.merge(students, grades, on="id")
print(df)
print(df.groupby("name")["mark"].agg(["mean", "max", "count"]))
print(df.pivot_table(index="name", columns="course", values="mark"))
```

```python
import pandas as pd

df = pd.DataFrame({
    "city": ["Tehran", "Shiraz", "Tehran", "Tabriz", "Shiraz"],
    "sales": [120, 85, None, 60, 95],
    "year": [2024, 2024, 2025, 2025, 2025],
})
print(df.isna().sum())
df["sales"] = df["sales"].fillna(df["sales"].mean())
print(df.head(3))
print(df[df.sales > 90])
print(df.groupby("city")["sales"].mean().round(1))
print(df.describe().round(1))
```
