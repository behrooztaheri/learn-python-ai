PyTorch کتابخانه اصلی یادگیری عمیق است. مشتق‌ها با **autograd** خودکار حساب می‌شوند.

حلقه آموزش چهار خط اصلی دارد: `zero_grad`، محاسبه خطا، `backward` و `step`.

> PyTorch در مرورگر اجرا نمی‌شود. دکمه Colab را بزن یا آن را با `pip install torch` روی کامپیوتر نصب کن.

```python
import torch, torch.nn as nn
from sklearn.datasets import load_digits

X, y = load_digits(return_X_y=True)
X = torch.tensor(X / 16, dtype=torch.float32)
y = torch.tensor(y)

model = nn.Sequential(nn.Linear(64, 128), nn.ReLU(), nn.Dropout(0.2), nn.Linear(128, 10))
loss_fn = nn.CrossEntropyLoss()
opt = torch.optim.Adam(model.parameters(), lr=1e-3)

for epoch in range(60):
    model.train()
    opt.zero_grad()
    loss = loss_fn(model(X[:1400]), y[:1400])
    loss.backward()
    opt.step()

model.eval()
with torch.no_grad():
    acc = (model(X[1400:]).argmax(1) == y[1400:]).float().mean()
print(f"Test accuracy: {acc:.3f}")
```
