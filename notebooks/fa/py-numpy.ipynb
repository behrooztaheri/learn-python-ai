NumPy آرایه چندبعدی و محاسبات سریع می‌دهد. عملیات روی کل آرایه بدون حلقه انجام می‌شود (**برداری‌سازی**) و **Broadcasting** آرایه‌های هم‌خوان با اندازه متفاوت را ترکیب می‌کند.

- `shape` ابعاد آرایه
- `axis=0` در امتداد سطرها (هر ستون) و `axis=1` در امتداد ستون‌ها (هر سطر)
- `@` ضرب ماتریسی

</div>

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.shape, a.mean(), a.mean(axis=0))
print(a * 10)
print(a[a > 2])
print(a @ a.T)

x = np.linspace(0, 1, 5)
print(x, x.reshape(5, 1).shape)
rng = np.random.default_rng(0)
print(rng.normal(size=3).round(3))
```
