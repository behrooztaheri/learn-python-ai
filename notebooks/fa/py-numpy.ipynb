# NumPy

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

NumPy آرایه چندبعدی و محاسبات سریع می‌دهد. عملیات روی کل آرایه بدون حلقه انجام می‌شود (**برداری‌سازی**) و **Broadcasting** آرایه‌های هم‌خوان با اندازه متفاوت را ترکیب می‌کند.

- `shape` ابعاد آرایه
- `axis=0` در امتداد سطرها (هر ستون) و `axis=1` در امتداد ستون‌ها (هر سطر)
- `@` ضرب ماتریسی

## ساختن آرایه
| تابع | نتیجه |
|---|---|
| `np.zeros((2,3))` | ماتریس صفر |
| `np.ones(4)` | بردار یک |
| `np.arange(0,10,2)` | گام مشخص |
| `np.linspace(0,1,5)` | تعداد نقطه مشخص |
| `np.eye(3)` | ماتریس همانی |

## Broadcasting در عمل
آرایه‌ای با ابعاد `(2,3)` را می‌توان با آرایه `(3,)` جمع کرد، چون بعد آخر برابر است و سطر کوچک‌تر تکرار می‌شود.



</div>

```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
print(A + np.array([10, 20, 30]))
print(A[:, 1], A[1, :], A[0, 2])

B = np.array([[2, 1], [1, 3]])
b = np.array([3, 5])
print("solve Bx=b:", np.linalg.solve(B, b))
print("inverse:\n", np.linalg.inv(B).round(2))
```

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.shape, a.mean(), a.mean(axis=0))
print(a * 10)
print(a[a > 2])
print(a @ a.T)

x = np.linspace(0, 1, 5)
print(x, x.reshape(5, 1).shape)
rng = np.random.default_rng(0)
print(rng.normal(size=3).round(3))
```
