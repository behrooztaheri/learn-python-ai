# کار با فایل

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

`with open` فایل را باز می‌کند و در پایان خودکار می‌بندد. حالت `w` می‌نویسد، `a` به انتها اضافه می‌کند و `r` می‌خواند. برای متن فارسی `encoding="utf-8"` را بگذار.

ماژول `pathlib` با مسیرها در همه سیستم‌عامل‌ها درست کار می‌کند و `json` داده ساختاریافته را ذخیره می‌کند.

</div>

```python
from pathlib import Path
import json

path = Path("notes.txt")
with open(path, "w", encoding="utf-8") as f:
    f.write("first line\nsecond line\n")

for line in path.read_text(encoding="utf-8").splitlines():
    print(line.upper())

data = {"course": "Python", "lessons": 20}
Path("data.json").write_text(json.dumps(data))
print(json.loads(Path("data.json").read_text())["lessons"])
```
