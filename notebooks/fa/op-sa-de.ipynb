# شبیه‌سازی تبرید و تکامل تفاضلی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

## شبیه‌سازی تبرید
حرکت بهتر همیشه پذیرفته می‌شود و حرکت بدتر با احتمال زیر؛ بنابراین می‌تواند از کمینه محلی فرار کند:

$$P=e^{-\Delta f/T}$$

دما $T$ به‌تدریج کم می‌شود.

| $\Delta f$ | T | P |
|---|---|---|
| 1 | 10 | 0.905 |
| 1 | 1 | 0.368 |
| 1 | 0.1 | 0.00005 |

## تکامل تفاضلی
فرزند از اختلاف اعضای جمعیت ساخته می‌شود: $v=x_a+F(x_b-x_c)$.



</div>

```python
import numpy as np
from scipy.optimize import differential_evolution

f = lambda x: 10 * len(x) + np.sum(x ** 2 - 10 * np.cos(2 * np.pi * x))
rng = np.random.default_rng(0)

x = rng.uniform(-5.12, 5.12, 2)
best, fbest, T = x.copy(), f(x), 10.0
for i in range(5000):
    y = np.clip(x + rng.normal(0, 0.5, 2), -5.12, 5.12)
    d = f(y) - f(x)
    if d < 0 or rng.random() < np.exp(-d / T):
        x = y
        if f(x) < fbest:
            best, fbest = x.copy(), f(x)
    T = max(T * 0.999, 1e-3)
print("SA:", best.round(3), round(fbest, 4))

r = differential_evolution(f, [(-5.12, 5.12)] * 2, seed=0)
print("DE:", r.x.round(4), round(r.fun, 6))
```
