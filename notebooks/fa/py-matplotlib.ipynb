# رسم نمودار با Matplotlib

درسی از [مدرسه پایتون و هوش مصنوعی](https://www.linkedin.com/in/behrooz-taheri-phd-smieee-b41b26113/)، دکتر بهروز طاهری

نمودار خوب الگوهایی را نشان می‌دهد که در جدول دیده نمی‌شوند. پرکاربردها: خطی برای روند، پراکندگی برای رابطه، هیستوگرام برای توزیع و میله‌ای برای مقایسه.

همیشه برای نمودار عنوان و نام محورها بگذار.

</div>

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(1)
x = rng.normal(170, 8, 300)
y = 0.9 * x - 90 + rng.normal(0, 6, 300)

fig, ax = plt.subplots(1, 2, figsize=(9, 3.5))
ax[0].hist(x, bins=25, color="#e8590c")
ax[0].set(title="Height distribution", xlabel="Height (cm)", ylabel="Count")
ax[1].scatter(x, y, s=10, color="#5b34c4")
ax[1].set(title="Height vs weight", xlabel="Height (cm)", ylabel="Weight (kg)")
plt.tight_layout()
plt.show()
```
