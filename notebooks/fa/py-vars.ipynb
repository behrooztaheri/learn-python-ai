# متغیرها و نوع داده

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

متغیر نامی است که به یک مقدار اشاره می‌کند. پایتون نوع را از روی مقدار تشخیص می‌دهد.

- `int` عدد صحیح
- `float` عدد اعشاری
- `str` متن
- `bool` درست یا نادرست

نام متغیر با عدد شروع نمی‌شود و به حروف بزرگ و کوچک حساس است.

## قواعد نام‌گذاری
| درست | نادرست | دلیل |
|---|---|---|
| `total_price` | `total price` | فاصله مجاز نیست |
| `score2` | `2score` | با عدد شروع نمی‌شود |
| `_count` | `class` | کلمه رزرو‌شده پایتون است |

قرارداد رایج پایتون برای نام متغیر، حروف کوچک همراه با خط زیر (`snake_case`) است.

## متغیر، برچسب است
در پایتون متغیر جعبه نیست؛ برچسبی است که به یک شیء اشاره می‌کند. با دستور زیر مقدار کپی نمی‌شود و دو نام به یک لیست اشاره می‌کنند.



</div>

```python
a = [1, 2]
b = a
b.append(3)
print(a, a is b)

c = a.copy()
c.append(4)
print(a, c, a is c)
```

## جدول تبدیل نوع
| ورودی | تبدیل | نتیجه |
|---|---|---|
| `"3.5"` | `float` | `3.5` |
| `3.9` | `int` | `3` (بخش اعشاری حذف می‌شود) |
| `0` | `bool` | `False` |
| `"abc"` | `int` | خطای `ValueError` |

</div>

```python
age = 25
height = 1.78
name = "Sara"
is_student = True

print(type(age), type(height), type(name), type(is_student))

# Converting between types
print(int("42") + 1)
print(str(3.5) + " meters")
```
