> [عملگرهای AND، OR و NOT.](https://raw.githubusercontent.com/behrooztaheri/learn-python-ai/main/assets/img/lessons/ops.png)

*عملگرهای AND، OR و NOT.*

```python
import numpy as np

def tri(x, a, b, c):
    return np.maximum(np.minimum((x - a) / (b - a), (c - x) / (c - b)), 0)

x = np.array([10, 20, 25, 30])
warm = tri(x, 15, 25, 35)
hot = tri(x, 25, 35, 45)
print("warm", warm.round(2))
print("hot ", hot.round(2))
print("warm AND hot", np.minimum(warm, hot).round(2))
print("warm OR hot ", np.maximum(warm, hot).round(2))
```
