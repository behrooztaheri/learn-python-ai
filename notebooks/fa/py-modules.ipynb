# ماژول و کتابخانه

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

هر فایل پایتون یک ماژول است. با `import` از کد آماده استفاده می‌کنیم. کتابخانه استاندارد همراه پایتون نصب می‌شود؛ مانند `math` و `random` و `datetime` و `collections`.

بلوک `if __name__ == "__main__":` فقط هنگام اجرای مستقیم فایل اجرا می‌شود، نه هنگام import.

</div>

```python
import math
import random
from collections import Counter
from datetime import date

print(math.sqrt(144), round(math.pi, 4))
random.seed(1)
print(random.randint(1, 6))
print(Counter("mississippi").most_common(2))
print(date(2026, 1, 1).strftime("%A"))
```
