تابع قطعه‌ای کد با نام است که ورودی می‌گیرد و با `return` نتیجه برمی‌گرداند. با تابع، کد تکرار نمی‌شود و خواناتر است.

ورودی‌ها می‌توانند مقدار پیش‌فرض داشته باشند. `lambda` تابعی کوتاه و بدون نام است.

## اجزای تابع
| بخش | معنی |
|---|---|
| پارامتر | نام ورودی در تعریف تابع |
| آرگومان | مقدار واقعی هنگام فراخوانی |
| مقدار پیش‌فرض | اگر آرگومان داده نشود |
| `*args` | تعداد دلخواه ورودی موقعیتی |
| `**kwargs` | تعداد دلخواه ورودی نام‌دار |

## حوزه متغیر
متغیری که داخل تابع ساخته می‌شود بیرون آن وجود ندارد. داده را با پارامتر بگیر و با `return` برگردان.

> مقدار پیش‌فرض را لیست نگذار. به‌جای `def f(a=[])` از `def f(a=None)` استفاده کن.



</div>

```python
def stats(*values, precision=2):
    mean = sum(values) / len(values)
    return round(mean, precision), max(values)

avg, top = stats(3, 7, 8, precision=1)
print(avg, top)

def describe(**info):
    for k, v in info.items():
        print(k, "->", v)

describe(name="Ali", city="Rasht")
```

```python
def bmi(weight, height=1.70):
    """Return body mass index."""
    return weight / height ** 2

print(round(bmi(70), 1))
print(round(bmi(70, height=1.85), 1))

people = [("Ali", 20), ("Sara", 18), ("Reza", 25)]
print(sorted(people, key=lambda p: p[1]))
```
