تابع قطعه‌ای کد با نام است که ورودی می‌گیرد و با `return` نتیجه برمی‌گرداند. با تابع، کد تکرار نمی‌شود و خواناتر است.

ورودی‌ها می‌توانند مقدار پیش‌فرض داشته باشند. `lambda` تابعی کوتاه و بدون نام است.

</div>

```python
def bmi(weight, height=1.70):
    """Return body mass index."""
    return weight / height ** 2

print(round(bmi(70), 1))
print(round(bmi(70, height=1.85), 1))

people = [("Ali", 20), ("Sara", 18), ("Reza", 25)]
print(sorted(people, key=lambda p: p[1]))
```
