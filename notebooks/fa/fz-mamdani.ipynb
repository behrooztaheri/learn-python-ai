قواعد به زبان ساده نوشته می‌شوند: «اگر هوا گرم است، آنگاه سرعت فن زیاد است.»

- **فازی‌سازی:** درجه عضویت ورودی‌ها
- **ارزیابی قواعد:** برش خروجی‌ها با قدرت قاعده
- **تجمیع:** ترکیب نتیجه‌ها با max
- **نافازی‌سازی:** تبدیل به عدد با مرکز ثقل

$$z^*=\frac{\int\mu(z)z\,dz}{\int\mu(z)\,dz}$$

</div>

```python
import numpy as np

def tri(x, a, b, c):
    return np.maximum(np.minimum((x - a) / (b - a + 1e-9), (c - x) / (c - b + 1e-9)), 0)

z = np.linspace(0, 100, 1001)
slow, medium, fast = tri(z, -1, 0, 50), tri(z, 0, 50, 100), tri(z, 50, 100, 101)

def fan_speed(t):
    cold, warm, hot = tri(t, 0, 10, 25), tri(t, 15, 25, 35), tri(t, 25, 40, 55)
    agg = np.maximum.reduce([np.minimum(cold, slow), np.minimum(warm, medium), np.minimum(hot, fast)])
    return (z * agg).sum() / agg.sum()

for t in [12.0, 22.0, 30.0, 40.0]:
    print(f"{t} C -> fan {fan_speed(t):.1f}%")
```
