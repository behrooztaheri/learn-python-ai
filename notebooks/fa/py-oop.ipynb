# کلاس و شیء

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

کلاس نقشه ساخت یک نوع شیء است و داده (ویژگی) و رفتار (متد) را کنار هم نگه می‌دارد. `__init__` هنگام ساخت شیء اجرا می‌شود و `self` به همان شیء اشاره می‌کند.

مدل‌های scikit-learn هم کلاس‌اند؛ برای همین می‌نویسیم `model.fit(X, y)`.

## چهار ایده اصلی
| ایده | معنی |
|---|---|
| کپسوله‌سازی | داده و رفتار کنار هم |
| وراثت | کلاس فرزند از کلاس والد استفاده می‌کند |
| چندریختی | متد هم‌نام، رفتار متفاوت |
| انتزاع | کاربر فقط رابط را می‌بیند |

متد `__str__` مشخص می‌کند شیء هنگام چاپ چگونه نمایش داده شود.



</div>

```python
class Shape:
    def area(self):
        raise NotImplementedError
    def __str__(self):
        return f"{type(self).__name__} area = {self.area():.2f}"

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14159 * self.r ** 2

class Square(Shape):
    def __init__(self, a):
        self.a = a
    def area(self):
        return self.a ** 2

for s in [Circle(1), Square(2)]:
    print(s)
```

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        self.balance += amount
        return self.balance

a = Account("Mina")
a.deposit(50)
print(a.owner, a.balance)

class Savings(Account):
    def add_interest(self, rate):
        return self.deposit(self.balance * rate)

s = Savings("Omid", 100)
print(s.add_interest(0.1))
```
