# گرادیان کاهشی

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

گرادیان کاهشی با گام‌های کوچک در خلاف جهت گرادیان، کمینه تابع هزینه را پیدا می‌کند.

$$\theta_{t+1}=\theta_t-\eta\nabla J(\theta_t)$$

نرخ یادگیری $\eta$ را تغییر بده: خیلی کوچک کند است و خیلی بزرگ واگرا می‌شود.

## سه نوع گرادیان کاهشی
| نوع | داده هر گام | مزیت | عیب |
|---|---|---|---|
| Batch | همه داده | پایدار | کند روی داده بزرگ |
| Stochastic | یک نمونه | بسیار سریع | پرنوسان |
| Mini-batch | چند ده نمونه | تعادل خوب | نیاز به تنظیم اندازه دسته |

## یک گام دستی
برای $J(x)=x^2$ با $x=4$ و $\eta=0.1$: گرادیان $2x=8$ است و $x_{new}=4-0.1\times8=3.2$.

| نرخ یادگیری | رفتار |
|---|---|
| خیلی کوچک | پیشرفت بسیار کند |
| مناسب | هم‌گرایی سریع |
| زیاد | نوسان |
| خیلی زیاد | واگرایی |

> همیشه ورودی‌ها را مقیاس‌بندی کن؛ در غیر این صورت سطح هزینه کشیده می‌شود و گرادیان کند حرکت می‌کند.

</div>

```python
import numpy as np

rng = np.random.default_rng(0)
x = rng.uniform(-1, 1, 100)
y = 3 * x + 2 + rng.normal(0, 0.2, 100)

w, b, lr = 0.0, 0.0, 0.1
for step in range(201):
    e = (w * x + b) - y
    w -= lr * 2 * np.mean(e * x)
    b -= lr * 2 * np.mean(e)
    if step % 50 == 0:
        print(step, round(np.mean(e ** 2), 4), round(w, 3), round(b, 3))
```
