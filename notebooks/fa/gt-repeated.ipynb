# بازی تکرارشونده و همکاری

درسی از مدرسه پایتون و هوش مصنوعی، دکتر بهروز طاهری

وقتی بازی تکرار شود، همکاری می‌تواند پایدار شود، چون خیانت امروز تنبیه فردا دارد. در مسابقه معروف آکسلرود، راهبرد ساده **Tit for Tat** بسیار موفق بود.

| راهبرد | قاعده |
|---|---|
| Always Cooperate | همیشه همکاری |
| Always Defect | همیشه خیانت |
| Tit for Tat | حرکت قبلی طرف مقابل را تکرار کن |
| Grim Trigger | پس از یک خیانت، برای همیشه خیانت |



</div>

```python
PAY = {("C", "C"): 3, ("C", "D"): 0, ("D", "C"): 5, ("D", "D"): 1}

def always_c(me, you): return "C"
def always_d(me, you): return "D"
def tit_for_tat(me, you): return you[-1] if you else "C"
def grim(me, you): return "D" if "D" in you else "C"

def play(f, g, rounds=100):
    a, b, sa, sb = [], [], 0, 0
    for _ in range(rounds):
        x, y = f(a, b), g(b, a)
        a.append(x); b.append(y)
        sa += PAY[(x, y)]; sb += PAY[(y, x)]
    return sa, sb

players = [always_c, always_d, tit_for_tat, grim]
total = {p.__name__: sum(play(p, q)[0] for q in players) for p in players}
print(sorted(total.items(), key=lambda kv: -kv[1]))
```
